Ya vimos los tipos simples de Python: ``int``, ``float``, ``complex``, ``bool``, ``str``, etc.
Python también tiene varios tipos compuestos integrados, que funcionan como contenedores de otros tipos.
Estos tipos compuestos son:

| Nombre del tipo | Ejemplo                   | Descripción                                    |
|-----------------|---------------------------|------------------------------------------------|
| ``list``        | ``[1, 2, 3]``             | Colección ordenada                             |
| ``tuple``       | ``(1, 2, 3)``             | Colección ordenada inmutable                   |
| ``dict``        | ``{'a':1, 'b':2, 'c':3}`` | Asociación de pares (clave, valor)             |
| ``set``         | ``{1, 2, 3}``             | Colección no ordenada de valores únicos        |

Como ves, los paréntesis, los corchetes y las llaves tienen significados distintos según el tipo de colección que se
crea. Haremos un recorrido rápido por estas estructuras de datos.

> 📝 **Nota de MiAtlas.cl:** el original describía el diccionario como una asociación «no ordenada». Desde

## Listas
Las listas son el tipo básico de colección de datos *ordenada* y *mutable* de Python.
Se definen con valores separados por comas entre corchetes; por ejemplo, esta es una lista de los primeros números
primos:

```python
L = [2, 3, 5, 7]
```

Las listas tienen varias propiedades y métodos útiles.
Veamos rápidamente algunos de los más comunes:

```python
# Largo de una lista
len(L)
```

```python
# Agregar un valor al final
L.append(11)
L
```

```python
# La suma concatena listas
L + [13, 17, 19]
```

```python
# El método sort() ordena la lista en el mismo lugar
L = [2, 5, 1, 6, 3, 4]
L.sort()
L
```

Además, hay muchos otros métodos de listas integrados; están bien explicados en la
[documentación en línea](https://docs.python.org/es/3/tutorial/datastructures.html) de Python.

Hasta ahora mostramos listas con valores de un solo tipo, pero una de las características más potentes de los objetos
compuestos de Python es que pueden contener objetos de *cualquier* tipo, o incluso una mezcla de tipos. Por ejemplo:

```python
L = [1, 'dos', 3.14, [0, 3, 5]]
```

Esta flexibilidad es consecuencia del sistema de tipos dinámico de Python.
¡Crear una secuencia mixta así en un lenguaje de tipado estático como C puede ser un gran dolor de cabeza!
Vemos que las listas pueden contener incluso otras listas como elementos.
Esta flexibilidad de tipos es una pieza esencial de lo que hace que el código de Python sea relativamente rápido y
fácil de escribir.

Hasta aquí hemos trabajado con las listas como un todo; otra pieza esencial es el acceso a los elementos
individuales. En Python esto se hace con la *indexación* y el *rebanado*, que veremos a continuación.

### Indexación y rebanado de listas
Python da acceso a los elementos de los tipos compuestos mediante la *indexación*, para elementos individuales, y el
*rebanado*, para varios elementos.
Como veremos, ambos se escriben con corchetes.
Volvamos a nuestra lista de los primeros números primos:

```python
L = [2, 3, 5, 7, 11]
```

Python usa indexación *desde cero*, así que accedemos al primer y al segundo elemento con esta sintaxis:

```python
L[0]
```

```python
L[1]
```

A los elementos del final de la lista se accede con números negativos, empezando por -1:

```python
L[-1]
```

```python
L[-2]
```

Puedes visualizar este esquema de indexación así:

![Figura de indexación de listas: los valores 2, 3, 5, 7 y 11 en casillas, con los índices positivos (0 a 5) arriba y los negativos (-5 a -1) abajo](https://raw.githubusercontent.com/jakevdp/WhirlwindTourOfPython/master/fig/list-indexing.png)

Aquí los valores de la lista se representan con los números grandes dentro de las casillas; los índices de la
lista, con los números pequeños de arriba y de abajo.
En este caso, ``L[2]`` devuelve ``5``, porque es el valor que sigue al índice ``2``.

Mientras que la *indexación* sirve para obtener un solo valor de la lista, el *rebanado* sirve para acceder a
varios valores en sublistas.
Usa dos puntos para indicar el punto de inicio (incluido) y el punto final (excluido) de la sublista.
Por ejemplo, para obtener los tres primeros elementos de la lista, escribimos:

```python
L[0:3]
```

Fíjate dónde quedan ``0`` y ``3`` en el diagrama anterior, y cómo el rebanado toma solo los valores que están
entre esos índices.
Si omitimos el primer índice, se asume ``0``, así que podemos escribir lo mismo así:

```python
L[:3]
```

Del mismo modo, si omitimos el último índice, se toma por defecto el largo de la lista.
Así, a los tres últimos elementos se accede de esta forma:

```python
L[-3:]
```

Por último, se puede indicar un tercer entero que representa el paso; por ejemplo, para seleccionar uno de
cada dos elementos de la lista, escribimos:

```python
L[::2]  # equivale a L[0:len(L):2]
```

Una versión especialmente útil es indicar un paso negativo, que invierte la lista:

```python
L[::-1]
```

Tanto la indexación como el rebanado sirven para asignar elementos, no solo para leerlos.
La sintaxis es la que esperarías:

```python
L[0] = 100
print(L)
```

```python
L[1:3] = [55, 56]
print(L)
```

Muchos paquetes orientados a la ciencia de datos, como NumPy y Pandas (mencionados en la introducción), usan
una sintaxis de rebanado muy parecida.

Ahora que vimos las listas de Python y cómo acceder a los elementos de los tipos compuestos ordenados, veamos los
otros tres tipos de datos compuestos estándar que mencionamos antes.

## Tuplas
En muchos aspectos, las tuplas se parecen a las listas, pero se definen con paréntesis en vez de corchetes:

```python
t = (1, 2, 3)
```

También se pueden definir sin ningún paréntesis:

```python
t = 1, 2, 3
print(t)
```

Igual que las listas, las tuplas tienen un largo, y sus elementos individuales se obtienen con indexación entre
corchetes:

```python
len(t)
```

```python
t[0]
```

La principal característica de las tuplas es que son *inmutables*: una vez creadas, no se puede cambiar ni
su tamaño ni su contenido (las dos celdas siguientes fallan a propósito):

```python
t[1] = 4
```

```python
t.append(4)
```

Las tuplas se usan mucho en los programas de Python; un caso muy común son las funciones que devuelven varios
valores.
Por ejemplo, el método ``as_integer_ratio()`` de los números de punto flotante devuelve un numerador y un
denominador; este doble valor devuelto viene en forma de tupla:

```python
x = 0.125
x.as_integer_ratio()
```

Estos valores devueltos se pueden asignar por separado así:

```python
numerador, denominador = x.as_integer_ratio()
print(numerador / denominador)
```

La lógica de indexación y rebanado que vimos para las listas también funciona con las tuplas, junto con otros
métodos. Consulta la [documentación de Python](https://docs.python.org/es/3/tutorial/datastructures.html) en línea
para ver una lista más completa.

## Diccionarios
Los diccionarios son asociaciones muy flexibles de claves a valores, y son la base de gran parte de la implementación
interna de Python.
Se crean con una lista de pares ``clave:valor`` separados por comas entre llaves:

```python
numeros = {'uno': 1, 'dos': 2, 'tres': 3}
```

Los elementos se leen y se asignan con la misma sintaxis de indexación de las listas y las tuplas, salvo que
aquí el índice no es una posición que parte en cero, sino una clave válida del diccionario:

```python
# Leer un valor a partir de su clave
numeros['dos']
```

También se pueden agregar elementos nuevos al diccionario con indexación:

```python
# Asignar un nuevo par clave:valor
numeros['noventa'] = 90
print(numeros)
```

Los diccionarios están implementados como *tablas hash* (si te da curiosidad cómo funcionan, lee sobre ese
concepto), lo que permite que el acceso a cualquier elemento sea muy rápido, sin importar el tamaño del diccionario.
La [documentación de Python](https://docs.python.org/es/3/library/stdtypes.html) tiene la lista completa de los
métodos disponibles para los diccionarios.

> 📝 **Nota de MiAtlas.cl:** el original (2016) decía aquí que los diccionarios *no* mantienen ningún orden.
> Eso cambió: desde **Python 3.7** el lenguaje garantiza que los diccionarios **conservan el orden de inserción**
> (en CPython 3.6 ya ocurría, pero como detalle de la implementación). Por eso ``print(numeros)`` muestra
> ``'noventa'`` al final: fue la última clave que agregaste. Aun así, a los valores se accede por clave, no por
> posición: ``numeros[0]`` da un ``KeyError``.

```python
# El orden de inserción se conserva al recorrer el diccionario
for clave, valor in numeros.items():
    print(clave, valor)
```

## Conjuntos

La cuarta colección básica es el conjunto, que contiene colecciones no ordenadas de elementos únicos.
Se definen de forma parecida a las listas y las tuplas, pero con las llaves de los diccionarios:

```python
primos = {2, 3, 5, 7}
impares = {1, 3, 5, 7, 9}
```

Si conoces la matemática de conjuntos, te resultarán familiares operaciones como la unión, la intersección, la
diferencia, la diferencia simétrica y otras.
Los conjuntos de Python traen todas estas operaciones, mediante métodos u operadores.
Para cada una mostraremos las dos formas equivalentes (la celda muestra el resultado de la última línea):

```python
# unión: elementos que aparecen en cualquiera de los dos
primos | impares      # con un operador
primos.union(impares) # lo mismo con un método
```

```python
# intersección: elementos que aparecen en ambos
primos & impares             # con un operador
primos.intersection(impares) # lo mismo con un método
```

```python
# diferencia: elementos de primos que no están en impares
primos - impares           # con un operador
primos.difference(impares) # lo mismo con un método
```

```python
# diferencia simétrica: elementos que aparecen en uno solo de los conjuntos
primos ^ impares                     # con un operador
primos.symmetric_difference(impares) # lo mismo con un método
```

Hay muchos más métodos y operaciones de conjuntos.
Seguro ya adivinas lo que voy a decir: consulta la [documentación en línea](https://docs.python.org/es/3/library/stdtypes.html)
de Python para tener la referencia completa.

## Estructuras de datos más especializadas

Python tiene otras estructuras de datos que pueden serte útiles; en general están en el módulo ``collections`` de la
biblioteca estándar.
El módulo ``collections`` está documentado por completo en la
[documentación en línea de Python](https://docs.python.org/es/3/library/collections.html), donde puedes leer más
sobre los distintos objetos disponibles.

En particular, a mí me han resultado muy útiles en ocasiones:

- ``collections.namedtuple``: como una tupla, pero cada valor tiene un nombre.
- ``collections.defaultdict``: como un diccionario, pero las claves que no existen tienen un valor por defecto que
  defines tú.
- ``collections.OrderedDict``: como un diccionario, pero conserva el orden de las claves.

Una vez que conoces los tipos de colecciones integrados estándar, usar estas funcionalidades extendidas es muy
intuitivo; te sugiero [leer sobre su uso](https://docs.python.org/es/3/library/collections.html).

> 📝 **Nota de MiAtlas.cl:** como el ``dict`` común ya conserva el orden de inserción (Python 3.7), hoy
> ``OrderedDict`` se usa poco: sirve cuando necesitas su método ``move_to_end()`` o que la igualdad tenga en cuenta
> el orden. Del mismo módulo, vale la pena conocer también ``Counter`` (cuenta elementos) y ``deque`` (cola con
> inserción rápida por ambos extremos). Un ejemplo breve:

```python
from collections import namedtuple, defaultdict, Counter, deque

Punto = namedtuple('Punto', ['x', 'y'])
p = Punto(3, 4)
print(p.x, p.y)

por_inicial = defaultdict(list)
for palabra in ['uva', 'pera', 'uvilla', 'palta']:
    por_inicial[palabra[0]].append(palabra)
print(dict(por_inicial))

print(Counter('abracadabra').most_common(2))

cola = deque([1, 2, 3])
cola.appendleft(0)
print(cola)
```

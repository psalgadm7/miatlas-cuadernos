## Manipulación simple de cadenas en Python

Para la manipulación básica de cadenas, los métodos incorporados de Python son muy cómodos.
Si vienes de C o de otro lenguaje de bajo nivel, probablemente su sencillez te parecerá un alivio.
Ya presentamos el tipo cadena de Python y algunos de estos métodos; aquí profundizaremos un poco más.

### Dar formato a cadenas: mayúsculas y minúsculas

Python facilita mucho cambiar entre mayúsculas y minúsculas.
Veremos los métodos ``upper()``, ``lower()``, ``capitalize()``, ``title()`` y ``swapcase()``, con esta cadena desordenada como ejemplo:

````python
zorro = "eL vELOz zORRo mARRóN."
````

Para pasar la cadena entera a mayúsculas o a minúsculas, usa los métodos ``upper()`` o ``lower()``, respectivamente:

````python
zorro.upper()
````

````python
zorro.lower()
````

Una necesidad frecuente es poner en mayúscula solo la primera letra de cada palabra, o quizás la primera letra de cada oración.
Eso se hace con los métodos ``title()`` y ``capitalize()``:

````python
zorro.title()
````

````python
zorro.capitalize()
````

Las mayúsculas y minúsculas se pueden intercambiar con el método ``swapcase()``:

````python
zorro.swapcase()
````

### Dar formato a cadenas: agregar y quitar espacios

Otra necesidad común es quitar espacios (u otros caracteres) del comienzo o del final de la cadena.
El método básico para quitar caracteres es ``strip()``, que elimina los espacios en blanco del comienzo y del final de la línea:

````python
linea = '         este es el contenido         '
linea.strip()
````

Para quitar los espacios solo a la derecha o a la izquierda, usa ``rstrip()`` o ``lstrip()``, respectivamente:

````python
linea.rstrip()
````

````python
linea.lstrip()
````

Para quitar caracteres distintos del espacio, puedes pasarle a ``strip()`` el carácter que quieras:

````python
num = "000000000000435"
num.strip('0')
````

> 📝 **Nota de MiAtlas.cl:** ``strip()`` recibe un *conjunto de caracteres*, no una palabra: ``'archivo.txt'.rstrip('.txt')`` quita cualquier ``.``, ``t`` o ``x`` del final. Para quitar un prefijo o un sufijo exacto, usa ``removeprefix()`` y ``removesuffix()`` (desde Python 3.9).

````python
print('texto.txt'.rstrip('.txt'))        # ¡se come también la «t» final de «texto»!
print('texto.txt'.removesuffix('.txt'))  # quita exactamente el sufijo
````

La operación opuesta, agregar espacios u otros caracteres, se logra con los métodos ``center()``, ``ljust()`` y ``rjust()``.

Por ejemplo, podemos usar ``center()`` para centrar una cadena dentro de cierta cantidad de espacios:

````python
linea = "este es el contenido"
linea.center(30)
````

De forma parecida, ``ljust()`` y ``rjust()`` alinean la cadena a la izquierda o a la derecha dentro de un largo dado:

````python
linea.ljust(30)
````

````python
linea.rjust(30)
````

Todos estos métodos aceptan además cualquier carácter para rellenar el espacio.
Por ejemplo:

````python
'435'.rjust(10, '0')
````

Como rellenar con ceros es algo tan común, Python ofrece también ``zfill()``, un método especial para rellenar una cadena con ceros a la izquierda:

````python
'435'.zfill(10)
````

### Buscar y reemplazar subcadenas

Si quieres encontrar dónde aparece cierto carácter en una cadena, los mejores métodos incorporados son ``find()``/``rfind()``, ``index()``/``rindex()`` y ``replace()``.

``find()`` e ``index()`` son muy parecidos: buscan la primera aparición de un carácter o subcadena dentro de una cadena y devuelven el índice donde está:

````python
linea = 'el veloz zorro marrón saltó sobre un perro perezoso'
linea.find('zorro')
````

````python
linea.index('zorro')
````

La única diferencia entre ``find()`` e ``index()`` es lo que hacen cuando no encuentran lo que buscan: ``find()`` devuelve ``-1``, mientras que ``index()`` lanza un ``ValueError``:

````python
linea.find('gato')
````

````python
linea.index('gato')
````

Los métodos relacionados ``rfind()`` y ``rindex()`` funcionan igual, salvo que buscan la primera aparición desde el final de la cadena, no desde el comienzo:

````python
linea.rfind('a')
````

Para el caso especial de revisar si una subcadena está al comienzo o al final de una cadena, Python ofrece los métodos ``startswith()`` y ``endswith()``:

````python
linea.endswith('perezoso')
````

````python
linea.startswith('zorro')
````

Para ir un paso más allá y reemplazar una subcadena por otra, puedes usar el método ``replace()``.
Reemplacemos ``'marrón'`` por ``'rojo'``:

````python
linea.replace('marrón', 'rojo')
````

La función ``replace()`` devuelve una cadena nueva y reemplaza todas las apariciones de lo que buscas:

````python
linea.replace('o', '--')
````

Para una versión más flexible de esta funcionalidad de ``replace()``, revisa más abajo la sección sobre expresiones regulares, «Búsqueda flexible de patrones con expresiones regulares».

### Dividir y partir cadenas

Si quieres encontrar una subcadena *y luego* dividir la cadena según su ubicación, los métodos que buscas son ``partition()`` y ``split()``.
Ambos devuelven una secuencia de subcadenas.

El método ``partition()`` devuelve una tupla de tres elementos: la subcadena anterior a la primera aparición del punto de corte, el punto de corte mismo y la subcadena posterior:

````python
linea.partition('zorro')
````

El método ``rpartition()`` es parecido, pero busca desde la derecha de la cadena.

El método ``split()`` es quizás más útil: encuentra *todas* las apariciones del punto de corte y devuelve las subcadenas que quedan entre ellas.
Por defecto divide en cualquier espacio en blanco, y devuelve una lista con las palabras de la cadena:

````python
linea.split()
````

Un método relacionado es ``splitlines()``, que divide en los saltos de línea.
Hagámoslo con un haiku, atribuido popularmente al poeta del siglo XVII Matsuo Bashō:

````python
haiku = """matsushima-ya
aah matsushima-ya
matsushima-ya"""

haiku.splitlines()
````

Si quieres deshacer un ``split()``, puedes usar el método ``join()``, que devuelve una cadena construida a partir de un separador y un iterable:

````python
'--'.join(['1', '2', '3'])
````

Un patrón común es usar el carácter especial ``"\n"`` (salto de línea) para unir líneas que antes se dividieron, y así recuperar el texto original:

````python
print("\n".join(['matsushima-ya', 'aah matsushima-ya', 'matsushima-ya']))
````

## Cadenas de formato

Con los métodos anteriores aprendimos a extraer valores de las cadenas y a darles a las cadenas mismas el formato que queremos.
Otro uso de los métodos de cadenas es manejar la *representación* como cadena de valores de otros tipos.
Por supuesto, la representación como cadena siempre se puede obtener con la función ``str()``; por ejemplo:

> 📝 **Nota de MiAtlas.cl:** al final de esta sección verás las ***f-strings***, la forma recomendada hoy para insertar valores en un texto.

````python
pi = 3.14159
str(pi)
````

Para formatos más complicados, quizás te tiente usar la aritmética de cadenas que vimos en «Semántica básica de Python: operadores»:

````python
"El valor de pi es " + str(pi)
````

Una forma más flexible de hacerlo es usar *cadenas de formato*, que son cadenas con marcadores especiales (indicados con llaves) donde se insertan los valores convertidos a cadena.
Este es un ejemplo básico:

````python
"El valor de pi es {}".format(pi)
````

Dentro del marcador ``{}`` también puedes indicar *qué* quieres que aparezca exactamente ahí.
Si pones un número, se refiere a la posición del argumento que se va a insertar:

````python
"""Primera letra: {0}. Última letra: {1}.""".format('A', 'Z')
````

Si pones un texto, se refiere al nombre de un argumento con nombre:

````python
"""Primera letra: {primera}. Última letra: {ultima}.""".format(ultima='Z', primera='A')
````

Por último, para valores numéricos puedes incluir códigos de formato que controlan cómo se convierte el valor en cadena.
Por ejemplo, para mostrar un número de punto flotante con tres dígitos después del punto decimal, puedes usar lo siguiente:

````python
"pi = {0:.3f}".format(pi)
````

Como antes, aquí el «``0``» se refiere a la posición del valor que se inserta.
Los «``:``» indican que vienen códigos de formato.
El «``.3f``» codifica la precisión deseada: tres dígitos después del punto decimal, en formato de punto flotante.

Este estilo de especificación de formato es muy flexible, y los ejemplos de aquí apenas rozan la superficie de las opciones disponibles.
Para más información sobre la sintaxis de estas cadenas de formato, revisa la sección [Especificación de formato](https://docs.python.org/es/3/library/string.html#formatspec) de la documentación de Python.

> 📝 **Nota de MiAtlas.cl:** desde Python 3.6, la forma **recomendada** de insertar valores en un texto son las ***f-strings*** (cadenas f): pones una ``f`` antes de las comillas y escribes la expresión entre llaves, directamente en el texto. Admiten los mismos códigos de formato que acabas de ver con ``.format()`` (por ejemplo, ``:.3f``), y algunos extras útiles:
>
> - ``{valor = }`` (desde 3.8) muestra la expresión junto a su valor, ideal para depurar;
> - ``:,`` agrega separador de miles y ``:.1%`` muestra un porcentaje;
> - desde Python 3.12 puedes repetir dentro de las llaves el mismo tipo de comillas de la cadena (PEP 701).
>
> El método ``.format()``, que el autor acaba de explicar, sigue funcionando y es útil cuando la plantilla se define antes de tener los valores (por ejemplo, si la lees de un archivo).

````python
f"El valor de pi es {pi}"
````

````python
primera, ultima = 'A', 'Z'
print(f"Primera letra: {primera}. Última letra: {ultima}.")
print(f"pi = {pi:.3f}")
print(f"{pi = :.2f}")

visitas = 1_234_567
print(f"Visitas: {visitas:,}")
print(f"Avance: {0.256:.1%}")

frutas = ['manzana', 'pera', 'uva']
print(f"Frutas: {", ".join(frutas)}")  # comillas repetidas: válido desde Python 3.12
````

## Búsqueda flexible de patrones con expresiones regulares

Los métodos del tipo ``str`` de Python te dan un conjunto poderoso de herramientas para dar formato, dividir y manipular datos de texto.
Pero hay herramientas aún más poderosas en el módulo incorporado de *expresiones regulares*.
Las expresiones regulares son un tema enorme; hay libros enteros sobre el tema (como [*Mastering Regular Expressions*, 3.ª edición](https://www.oreilly.com/library/view/mastering-regular-expressions/0596528124/), de Jeffrey E. F. Friedl), así que será difícil hacerle justicia en una sola subsección.

Mi objetivo aquí es darte una idea de los problemas que se pueden abordar con expresiones regulares, y una idea básica de cómo usarlas en Python.
Al final sugiero algunas referencias para aprender más, en «Más recursos sobre expresiones regulares».

En el fondo, las expresiones regulares son una forma de *buscar patrones de manera flexible* en cadenas.
Si usas la línea de comandos a menudo, probablemente conoces este tipo de búsqueda flexible con el carácter «``*``», que actúa como comodín.
Por ejemplo, podemos listar todos los cuadernos de Jupyter (es decir, los archivos con extensión *.ipynb*) que tengan «Python» en su nombre usando el comodín «``*``» para que coincida con cualquier cosa entremedio:

````python
import fnmatch

archivos = ['01-Python-basico.ipynb', 'datos.csv', 'Python-avanzado.ipynb',
            'notas.txt', 'Introduccion-a-Python.ipynb']
fnmatch.filter(archivos, '*Python*.ipynb')
````

Las expresiones regulares generalizan esta idea de «comodín» a una gran variedad de sintaxis flexibles para buscar coincidencias en cadenas.
La interfaz de Python para las expresiones regulares está en el módulo incorporado ``re``; como ejemplo sencillo, usémoslo para reproducir lo que hace el método ``split()`` de las cadenas:

````python
import re
regex = re.compile(r'\s+')
regex.split(linea)
````

> 📝 **Nota de MiAtlas.cl:** el original escribía ``re.compile('\s+')``, sin la ``r`` delante. Desde Python 3.12, una barra invertida que no forma una secuencia de escape válida (como ``\s`` o ``\w`` en una cadena normal) produce un ``SyntaxWarning``. Por eso, en este cuaderno todas las expresiones regulares se escriben como cadenas crudas (``r'...'``); más abajo se explica qué son.

Aquí primero *compilamos* una expresión regular y después la usamos para *dividir* una cadena.
Igual que el método ``split()`` de Python devuelve una lista con todas las subcadenas que quedan entre los espacios en blanco, el método ``split()`` de la expresión regular devuelve una lista con todas las subcadenas que quedan entre las coincidencias con el patrón.

En este caso, el patrón es ``"\s+"``: «``\s``» es un carácter especial que coincide con cualquier espacio en blanco (espacio, tabulación, salto de línea, etc.), y el «``+``» indica *una o más* repeticiones de lo que lo precede.
Así, la expresión regular coincide con cualquier subcadena formada por uno o más espacios.

El método ``split()`` es, en el fondo, una función de conveniencia construida sobre este comportamiento de *búsqueda de patrones*; más fundamental es el método ``match()``, que te dice si el comienzo de una cadena coincide con el patrón:

````python
for s in ["     ", "abc  ", "  abc"]:
    if regex.match(s):
        print(repr(s), "coincide")
    else:
        print(repr(s), "no coincide")
````

Igual que con ``split()``, hay funciones de conveniencia parecidas para encontrar la primera coincidencia (como ``str.index()`` o ``str.find()``) o para buscar y reemplazar (como ``str.replace()``).
Usaremos otra vez la línea de antes:

````python
linea = 'el veloz zorro marrón saltó sobre un perro perezoso'
````

Con ella vemos que el método ``regex.search()`` funciona de forma muy parecida a ``str.index()`` o ``str.find()``:

````python
linea.index('zorro')
````

````python
regex = re.compile('zorro')
coincidencia = regex.search(linea)
coincidencia.start()
````

Del mismo modo, el método ``regex.sub()`` funciona de forma muy parecida a ``str.replace()``:

````python
linea.replace('zorro', 'OSO')
````

````python
regex.sub('OSO', linea)
````

### Un ejemplo más elaborado

Pero, podrías preguntarte, ¿para qué usar la sintaxis más complicada y extensa de las expresiones regulares en vez de los métodos de cadenas, más intuitivos y simples?
La ventaja es que las expresiones regulares ofrecen *mucha* más flexibilidad.

Veamos un ejemplo más complicado: la tarea común de reconocer direcciones de correo electrónico.
Empezaré escribiendo una expresión regular (algo indescifrable) y después explicaré qué está pasando.
Aquí va:

````python
correo = re.compile(r'\w+@\w+\.[a-z]{3}')
````

Con ella, si nos dan una línea de un documento, podemos extraer rápidamente lo que parezcan direcciones de correo:

````python
texto = "Para escribirle a Guido, prueba guido@python.org o la dirección antigua guido@google.com."
correo.findall(texto)
````

(Ten en cuenta que estas direcciones son totalmente inventadas; probablemente hay mejores formas de contactar a Guido).

Podemos hacer más operaciones, como reemplazar las direcciones por otra cadena, quizás para ocultarlas en la salida:

````python
correo.sub('--@--.--', texto)
````

Por último, ten en cuenta que si de verdad quieres reconocer *cualquier* dirección de correo, la expresión regular anterior es demasiado simple.
Por ejemplo, solo admite direcciones formadas por caracteres alfanuméricos que terminan en un sufijo de dominio de tres letras.
Así, por ejemplo, el punto que aparece aquí hace que solo encontremos una parte de la dirección:

````python
correo.findall('ana.perez@ejemplo.com')
````

¡Esto muestra lo implacables que pueden ser las expresiones regulares si no tienes cuidado!
Si buscas en internet, encontrarás propuestas de expresiones regulares que reconocen *todos* los correos válidos, pero ojo: son mucho más complicadas que la expresión simple que usamos aquí.

### Lo básico de la sintaxis de las expresiones regulares

La sintaxis de las expresiones regulares es un tema demasiado grande para esta breve sección.
Aun así, conocerla un poco ayuda mucho: recorreré aquí algunas construcciones básicas y después listaré recursos más completos para aprender más.
Espero que esta introducción rápida te permita aprovechar bien esos recursos.

#### Las cadenas simples coinciden directamente

Si construyes una expresión regular con una cadena simple de caracteres o dígitos, coincidirá con esa cadena exacta:

````python
regex = re.compile('ción')
regex.findall('Emoción y acción')
````

#### Algunos caracteres tienen significados especiales

Aunque las letras y los números simples coinciden directamente, hay un puñado de caracteres que tienen significados especiales dentro de las expresiones regulares. Son estos:
```
. ^ $ * + ? { } [ ] \ | ( )
```
Veremos el significado de algunos de ellos en un momento.
Mientras tanto, debes saber que si quieres buscar cualquiera de estos caracteres tal cual, puedes *escaparlo* con una barra invertida:

````python
regex = re.compile(r'\$')
regex.findall("el precio es $20")
````

La ``r`` delante de ``r'\$'`` indica una *cadena cruda* (*raw string*); en las cadenas normales de Python, la barra invertida se usa para indicar caracteres especiales.
Por ejemplo, una tabulación se indica con ``"\t"``:

````python
print('a\tb\tc')
````

En una cadena cruda no se hacen esas sustituciones:

````python
print(r'a\tb\tc')
````

Por eso, siempre que uses barras invertidas en una expresión regular, es una buena práctica usar una cadena cruda.

#### Los caracteres especiales pueden coincidir con grupos de caracteres

Así como el carácter ``"\"`` dentro de una expresión regular puede escapar caracteres especiales y convertirlos en normales, también puede dar un significado especial a caracteres normales.
Estos caracteres especiales coinciden con grupos determinados de caracteres, y ya los vimos antes.
En la expresión regular de las direcciones de correo usamos ``"\w"``, un marcador especial que coincide con *cualquier carácter alfanumérico*. De la misma forma, en el ejemplo simple de ``split()`` vimos ``"\s"``, un marcador especial que indica *cualquier espacio en blanco*.

Combinándolos, podemos crear una expresión regular que coincida con *dos letras o dígitos cualesquiera separados por un espacio en blanco*:

````python
regex = re.compile(r'\w\s\w')
regex.findall('el zorro tiene 9 años')
````

Este ejemplo empieza a mostrar el poder y la flexibilidad de las expresiones regulares.

La siguiente tabla muestra algunos de estos caracteres que suelen ser útiles:

| Carácter | Descripción | Carácter | Descripción |
|---|---|---|---|
| ``"\d"`` | Coincide con cualquier dígito | ``"\D"`` | Coincide con cualquier carácter que no sea dígito |
| ``"\s"`` | Coincide con cualquier espacio en blanco | ``"\S"`` | Coincide con cualquier carácter que no sea espacio en blanco |
| ``"\w"`` | Coincide con cualquier carácter alfanumérico | ``"\W"`` | Coincide con cualquier carácter no alfanumérico |

Esta *no* es una lista ni una descripción completa; para más detalles, revisa la [documentación de la sintaxis de expresiones regulares](https://docs.python.org/es/3/library/re.html#regular-expression-syntax) de Python.

> 📝 **Nota de MiAtlas.cl:** en Python 3, con cadenas de texto, ``\w`` reconoce letras de cualquier idioma (``ñ``, ``á``, ``ü``…), dígitos y el guion bajo ``_``; ``\d`` reconoce también dígitos de otros sistemas de escritura. Si solo quieres caracteres ASCII, agrega la opción ``re.ASCII``.

````python
print(re.findall(r'\w+', 'año, pingüino y canción'))
print(re.findall(r'\w+', 'año, pingüino y canción', flags=re.ASCII))
````

#### Los corchetes coinciden con grupos de caracteres a tu medida

Si los grupos de caracteres incorporados no son lo bastante específicos, puedes usar corchetes para indicar cualquier conjunto de caracteres que te interese.
Por ejemplo, lo siguiente coincide con cualquier vocal minúscula:

````python
regex = re.compile('[aeiou]')
regex.split('consecuencia')
````

De forma parecida, puedes usar un guion para indicar un rango: por ejemplo, ``"[a-z]"`` coincide con cualquier letra minúscula, y ``"[1-3]"`` con cualquiera de ``"1"``, ``"2"`` o ``"3"``.
Por ejemplo, quizás necesites extraer de un documento ciertos códigos formados por una letra mayúscula seguida de un dígito. Podrías hacerlo así:

````python
regex = re.compile('[A-Z][0-9]')
regex.findall('1043879, G2, H6')
````

#### Los comodines coinciden con caracteres repetidos

Si quieres buscar una cadena con, digamos, tres caracteres alfanuméricos seguidos, puedes escribir, por ejemplo, ``"\w\w\w"``.
Como esto se necesita muy a menudo, hay una sintaxis específica para buscar repeticiones: llaves con un número:

````python
regex = re.compile(r'\w{3}')
regex.findall('El veloz zorro marrón')
````

También hay marcadores para buscar cualquier cantidad de repeticiones; por ejemplo, el carácter ``"+"`` coincide con *una o más* repeticiones de lo que lo precede:

````python
regex = re.compile(r'\w+')
regex.findall('El veloz zorro marrón')
````

Esta es una tabla de los marcadores de repetición que puedes usar en las expresiones regulares:

| Carácter | Descripción | Ejemplo |
|-----------|-------------|---------|
| ``?`` | Coincide con cero o una repetición de lo anterior | ``"ab?"`` coincide con ``"a"`` o ``"ab"`` |
| ``*`` | Coincide con cero o más repeticiones de lo anterior | ``"ab*"`` coincide con ``"a"``, ``"ab"``, ``"abb"``, ``"abbb"``… |
| ``+`` | Coincide con una o más repeticiones de lo anterior | ``"ab+"`` coincide con ``"ab"``, ``"abb"``, ``"abbb"``…, pero no con ``"a"`` |
| ``{n}`` | Coincide con ``n`` repeticiones de lo anterior | ``"ab{2}"`` coincide con ``"abb"`` |
| ``{m,n}`` | Coincide con entre ``m`` y ``n`` repeticiones de lo anterior | ``"ab{2,3}"`` coincide con ``"abb"`` o ``"abbb"`` |

Con estas bases en mente, volvamos a nuestro buscador de direcciones de correo:

````python
correo = re.compile(r'\w+@\w+\.[a-z]{3}')
````

Ahora podemos entender qué significa: queremos uno o más caracteres alfanuméricos (``"\w+"``), seguidos de la *arroba* (``"@"``), seguida de uno o más caracteres alfanuméricos (``"\w+"``), seguidos de un punto (``"\."``; fíjate en que hace falta escaparlo con una barra invertida), seguido de exactamente tres letras minúsculas.

Si ahora queremos modificarla para que la dirección con punto también coincida, podemos usar la notación de corchetes:

````python
correo2 = re.compile(r'[\w.]+@\w+\.[a-z]{3}')
correo2.findall('ana.perez@ejemplo.com')
````

Cambiamos ``"\w+"`` por ``"[\w.]+"``, así que coincidirá con cualquier carácter alfanumérico *o* con un punto.
Con esta expresión más flexible podemos reconocer una variedad más amplia de direcciones de correo (aunque todavía no todas: ¿se te ocurren otras limitaciones de esta expresión?).

#### Los paréntesis indican *grupos* que se extraen

En expresiones regulares compuestas, como nuestro buscador de correos, a menudo queremos extraer sus componentes en vez de la coincidencia completa. Eso se hace con paréntesis, que *agrupan* los resultados:

````python
correo3 = re.compile(r'([\w.]+)@(\w+)\.([a-z]{3})')
````

````python
texto = "Para escribirle a Guido, prueba guido@python.org o la dirección antigua guido@google.com."
correo3.findall(texto)
````

Como vemos, la agrupación extrae una lista con los componentes de cada dirección de correo.

Podemos ir un poco más allá y darles *nombre* a los componentes extraídos con la sintaxis ``"(?P<nombre> )"``; en ese caso, los grupos se pueden obtener como un diccionario de Python:

````python
correo4 = re.compile(r'(?P<usuario>[\w.]+)@(?P<dominio>\w+)\.(?P<sufijo>[a-z]{3})')
coincidencia = correo4.match('guido@python.org')
coincidencia.groupdict()
````

Combinar estas ideas (y parte de la potente sintaxis de las expresiones regulares que no vimos aquí) te permite extraer información de cadenas en Python de forma flexible y rápida.

> 📝 **Nota de MiAtlas.cl:** dos funciones del módulo ``re`` que vale la pena conocer: ``re.fullmatch()`` exige que el patrón coincida con la cadena **completa** (útil para validar un dato), y las funciones de nivel de módulo (``re.findall(patron, texto)``, ``re.sub(...)``…) sirven sin compilar antes el patrón: Python guarda en caché los patrones usados hace poco.

````python
print(re.fullmatch(r'\d{4}', '2026') is not None)   # True: son exactamente 4 dígitos
print(re.fullmatch(r'\d{4}', '2026-11') is not None)  # False: sobra texto
````

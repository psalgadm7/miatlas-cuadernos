Si lees suficiente código Python, tarde o temprano te toparás con una construcción concisa y eficiente llamada
*comprensión de listas*.
Es una característica de Python de la que probablemente te enamorarás si no la has usado antes; se ve más o menos así:

````python
[i for i in range(20) if i % 3 > 0]
````

El resultado es una lista de números que excluye los múltiplos de 3.
Aunque este ejemplo puede parecer algo confuso al principio, a medida que te familiarices con Python, leer y escribir
comprensiones de listas te resultará natural.

## Comprensiones de listas básicas

Las comprensiones de listas son simplemente una forma de comprimir en una sola línea, corta y legible, un bucle `for`
que construye una lista.
Por ejemplo, este es un bucle que construye una lista con los primeros 12 cuadrados de enteros:

````python
L = []
for n in range(12):
    L.append(n ** 2)
L
````

La comprensión de listas equivalente es esta:

````python
[n ** 2 for n in range(12)]
````

Como con muchas instrucciones de Python, casi puedes leer su significado en lenguaje natural: «construye una lista con
el cuadrado de `n` para cada `n` menor que 12».

La sintaxis básica, entonces, es `[`*`expr`* `for` *`var`* `in` *`iterable`*`]`, donde *`expr`* es cualquier
expresión válida, *`var`* es un nombre de variable e *`iterable`* es cualquier objeto iterable de Python.

> 📝 **Nota de MiAtlas.cl:** la variable de una comprensión vive solo dentro de ella: después de
> `[n ** 2 for n in range(12)]`, la `n` no queda definida fuera (a diferencia de la variable de un bucle `for`).

## Iteración múltiple

A veces quieres construir una lista no a partir de un valor, sino de dos. Para eso, basta con agregar otra
expresión `for` en la comprensión:

````python
[(i, j) for i in range(2) for j in range(3)]
````

Fíjate en que la segunda expresión `for` actúa como el índice interior, el que cambia más rápido en la lista
resultante.
Este tipo de construcción se puede extender a tres, cuatro o más iteradores dentro de la comprensión, aunque en algún
momento la legibilidad del código se resentirá.

## Condiciones sobre el iterador

Puedes controlar aún más la iteración agregando una condición al final de la expresión.
En el primer ejemplo de la sección recorrimos todos los números de 0 a 19, pero dejamos fuera los múltiplos de 3.
Míralo otra vez y fíjate en la construcción:

````python
[val for val in range(20) if val % 3 > 0]
````

La expresión `(val % 3 > 0)` resulta `True` salvo que `val` sea divisible por 3.
De nuevo, el significado se puede leer de inmediato en lenguaje natural: «construye una lista con cada valor menor que
20, pero solo si el valor no es divisible por 3».
Cuando te acostumbras, esto es mucho más fácil de escribir (y de entender de un vistazo) que la sintaxis equivalente
con un bucle:

````python
L = []
for val in range(20):
    if val % 3:
        L.append(val)
L
````

## Condiciones sobre el valor

Si has programado en C, quizás conozcas el condicional de una línea que permite el operador `?`:

```c
int absval = (val < 0) ? -val : val
```

Python tiene algo muy parecido, que se usa sobre todo dentro de comprensiones de listas, funciones `lambda` y otros
lugares donde se necesita una expresión simple:

````python
val = -10
val if val >= 0 else -val
````

Vemos que esto simplemente repite lo que hace la función incorporada `abs()`, pero la construcción permite hacer cosas
muy interesantes dentro de las comprensiones de listas.
Esto ya se está poniendo complejo, pero podrías hacer algo así:

````python
[val if val % 2 else -val
 for val in range(20) if val % 3]
````

Fíjate en el salto de línea dentro de la comprensión de listas, antes de la expresión `for`: es válido en Python, y
suele ser una buena forma de dividir comprensiones largas para que se lean mejor.
Revísalo con calma: lo que hacemos es construir una lista dejando fuera los múltiplos de 3 y cambiando el signo de
todos los múltiplos de 2.

Una vez que entiendes cómo funcionan las comprensiones de listas, es fácil pasar a otros tipos de comprensiones. La
sintaxis es casi la misma; lo único que cambia es el tipo de paréntesis que usas.

Por ejemplo, con llaves puedes crear un conjunto (`set`) con una *comprensión de conjuntos*:

````python
{n**2 for n in range(12)}
````

Recuerda que un conjunto es una colección sin duplicados.
La comprensión de conjuntos respeta esa regla y elimina los elementos repetidos:

````python
{a % 3 for a in range(1000)}
````

Con un pequeño cambio, agregando dos puntos (`:`), creas una *comprensión de diccionarios*:

````python
{n: n**2 for n in range(6)}
````

> 📝 **Nota de MiAtlas.cl:** los diccionarios conservan el orden de inserción desde Python 3.7, así que las claves de
> una comprensión de diccionarios quedan en el mismo orden en que se recorrieron. Los conjuntos, en cambio, no
> garantizan ningún orden.

Por último, si usas paréntesis en lugar de corchetes, obtienes lo que se llama una *expresión generadora*:

````python
(n**2 for n in range(12))
````

Una expresión generadora es, en esencia, una comprensión de listas en la que los elementos se generan a medida que se
necesitan y no todos de una vez. La simplicidad de esto esconde la potencia de esta característica del lenguaje: la
exploraremos más en el próximo capítulo, *Generadores*.

> 📝 **Nota de MiAtlas.cl:** cuando una expresión generadora es el único argumento de una función, puedes omitir los
> paréntesis extra: `sum(n**2 for n in range(12))` suma los cuadrados sin crear nunca la lista completa.

## 1. Comandos de la terminal con «!»

Una línea que empieza con `!` se ejecuta en la terminal de la máquina, no en Python.

```python
!pwd
!ls -la
!df -h / | tail -1
```

```python
# Una variable de Python entra a un comando con {llaves}
carpeta = "/"
!ls {carpeta} | head -5
```

## 2. Instalar paquetes

Colab trae muchos paquetes instalados. Para agregar uno, usa `%pip install`, que instala en el mismo Python del
cuaderno. **Fija la versión** (`==`) para que el cuaderno funcione igual mañana.

> Lo instalado **se pierde** cuando la máquina se borra: deja las instalaciones en las primeras celdas.

```python
%pip install -q tabulate==0.9.0
```

```python
from tabulate import tabulate

print(tabulate([["Colab", "GPU gratis"], ["JupyterLite", "en tu navegador"]], headers=["Herramienta", "Ventaja"]))
```

## 3. Las «magias» de IPython

Empiezan con `%` (una línea) o `%%` (toda la celda).

```python
%time suma = sum(range(10_000_000))
```

```python
%%writefile hola.py
print("Hola desde un archivo escrito por el cuaderno")
```

```python
!python hola.py
```

## Pruébalo tú

Con `%%writefile`, crea `cuadrados.py`, que imprima los cuadrados del 1 al 5 en una línea (`1 4 9 16 25`). Después
ejecútalo con `!python cuadrados.py` y guarda su salida con `salida = !python cuadrados.py`.

```python
salida = None   # ← crea el archivo en otra celda y luego: salida = !python cuadrados.py
```

```python
# Comprobación
if salida is None:
    print("Aún no lo resuelves.")
elif " ".join(salida).split() == ["1", "4", "9", "16", "25"]:
    print("✅ ¡Correcto!")
else:
    print("❌ La salida fue:", list(salida))
```

## Resumen

- `!comando` usa la terminal; `{variable}` pasa valores de Python.
- `%pip install paquete==versión` instala en el cuaderno; se pierde con la máquina.
- `%time`, `%%writefile` y otras magias vienen de IPython.

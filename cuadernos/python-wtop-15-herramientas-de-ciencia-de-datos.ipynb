## NumPy: Python numérico

NumPy ofrece una forma eficiente de guardar y manipular arreglos densos multidimensionales en Python.
Sus características importantes son:

- Ofrece la estructura ``ndarray``, que permite guardar y manipular de forma eficiente vectores, matrices y conjuntos de datos de más dimensiones.
- Ofrece una sintaxis legible y eficiente para operar con estos datos, desde la aritmética simple elemento por elemento hasta operaciones de álgebra lineal más complicadas.

En el caso más simple, los arreglos de NumPy se parecen mucho a las listas de Python.
Por ejemplo, este es un arreglo con los números del 1 al 9 (compáralo con la función incorporada ``range()`` de Python):

````python
import numpy as np
x = np.arange(1, 10)
x
````

Los arreglos de NumPy ofrecen tanto un almacenamiento eficiente de los datos como operaciones eficientes elemento por elemento.
Por ejemplo, para elevar al cuadrado cada elemento del arreglo, podemos aplicarle directamente el operador «``**``»:

````python
x ** 2
````

Compáralo con la comprensión de listas al estilo de Python, mucho más extensa, que da el mismo resultado:

````python
[val ** 2 for val in range(1, 10)]
````

A diferencia de las listas de Python (que tienen una sola dimensión), los arreglos de NumPy pueden ser multidimensionales.
Por ejemplo, aquí cambiamos la forma de nuestro arreglo ``x`` a un arreglo de 3x3:

````python
M = x.reshape((3, 3))
M
````

Un arreglo bidimensional es una forma de representar una matriz, y NumPy sabe hacer de forma eficiente las operaciones típicas con matrices. Por ejemplo, puedes calcular la traspuesta con ``.T``:

````python
M.T
````

o un producto matriz-vector con ``np.dot``:

````python
np.dot(M, [5, 6, 7])
````

> 📝 **Nota de MiAtlas.cl:** desde Python 3.5 existe el operador ``@`` para la multiplicación de matrices; con NumPy, ``M @ v`` da el mismo resultado que ``np.dot(M, v)`` y se lee más parecido a la notación matemática.

````python
M @ np.array([5, 6, 7])
````

e incluso operaciones más sofisticadas, como el cálculo de valores propios:

````python
np.linalg.eigvals(M)
````

Este tipo de manipulación de álgebra lineal está en la base de gran parte del análisis de datos moderno, sobre todo en los campos del aprendizaje automático (*machine learning*) y la minería de datos.

Para más información sobre NumPy, revisa «Recursos para seguir aprendiendo» (capítulo 16).

> 📝 **Nota de MiAtlas.cl:** desde NumPy 2.0, los valores sueltos se muestran con su tipo, por ejemplo ``np.int64(285)`` en lugar de ``285``. Es el mismo número; ``print()`` lo muestra sin el tipo.

## Pandas: datos en columnas con etiquetas

Pandas es un paquete mucho más nuevo que NumPy, y de hecho está construido sobre él.
Lo que ofrece Pandas es una interfaz con etiquetas para datos multidimensionales, en forma de un objeto ``DataFrame`` que resultará muy familiar a quienes usan R y lenguajes parecidos.
Los ``DataFrame`` de Pandas se ven más o menos así:

````python
import pandas as pd
df = pd.DataFrame({'etiqueta': ['A', 'B', 'C', 'A', 'B', 'C'],
                   'valor': [1, 2, 3, 4, 5, 6]})
df
````

La interfaz de Pandas permite, por ejemplo, seleccionar columnas por su nombre:

````python
df['etiqueta']
````

Aplicar operaciones de cadenas a todos los textos:

````python
df['etiqueta'].str.lower()
````

Aplicar agregaciones a los valores numéricos:

````python
df['valor'].sum()
````

Y, quizás lo más importante, hacer uniones y agrupaciones eficientes al estilo de las bases de datos:

````python
df.groupby('etiqueta').sum()
````

Aquí, en una sola línea, calculamos la suma de todos los objetos que comparten la misma etiqueta, algo mucho más extenso (y mucho menos eficiente) con las herramientas de NumPy y de Python puro.

Para más información sobre cómo usar Pandas, revisa «Recursos para seguir aprendiendo» (capítulo 16).

## SciPy: Python científico

SciPy es una colección de funciones científicas construida sobre NumPy.
El paquete empezó como un conjunto de envoltorios de Python para bibliotecas de Fortran muy conocidas de cálculo numérico, y ha crecido desde ahí.
Está organizado como un conjunto de submódulos, cada uno de los cuales implementa una clase de algoritmos numéricos.
Esta es una muestra incompleta de algunos de los más importantes para la ciencia de datos:

- ``scipy.fftpack``: transformadas rápidas de Fourier
- ``scipy.integrate``: integración numérica
- ``scipy.interpolate``: interpolación numérica
- ``scipy.linalg``: rutinas de álgebra lineal
- ``scipy.optimize``: optimización numérica de funciones
- ``scipy.sparse``: almacenamiento de matrices dispersas y álgebra lineal con ellas
- ``scipy.stats``: rutinas de análisis estadístico

Por ejemplo, veamos cómo interpolar una curva suave entre algunos datos:

> 📝 **Nota de MiAtlas.cl:** hoy SciPy recomienda ``scipy.fft`` en lugar de ``scipy.fftpack``, que quedó como interfaz heredada. Lo mismo pasa con ``interpolate.interp1d``, que usa el ejemplo siguiente: sigue funcionando, pero para código nuevo la documentación de SciPy sugiere otras funciones, como ``CubicSpline`` (lo verás justo después).

````python
from scipy import interpolate

# elegir ocho puntos entre 0 y 10
x = np.linspace(0, 10, 8)
y = np.sin(x)

# crear una función de interpolación cúbica
func = interpolate.interp1d(x, y, kind='cubic')

# interpolar en una grilla de 1.000 puntos
x_interp = np.linspace(0, 10, 1000)
y_interp = func(x_interp)

# graficar los resultados
plt.figure()  # figura nueva
plt.plot(x, y, 'o')
plt.plot(x_interp, y_interp);
````

Lo que vemos es una interpolación suave entre los puntos.

> 📝 **Nota de MiAtlas.cl:** la misma idea con ``CubicSpline``, la alternativa recomendada para código nuevo:

````python
from scipy.interpolate import CubicSpline

spline = CubicSpline(x, y)

plt.figure()
plt.plot(x, y, 'o')
plt.plot(x_interp, spline(x_interp));
````

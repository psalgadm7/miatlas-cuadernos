**Objetivo:** subir y bajar archivos, y guardar lo importante en **Google Drive** para que no se pierda.

La máquina de Colab es **temporal**: según la FAQ, se borra tras un rato sin uso y tiene una vida máxima. Todo lo que
escribas en su disco (la carpeta `/content`) desaparece con ella.

```python
# ¿Estamos en Colab? Los cuadernos funcionan también fuera de Colab, con avisos en lo que es solo de Colab.
try:
    import google.colab  # noqa: F401
    EN_COLAB = True
except ImportError:
    EN_COLAB = False
print("Corriendo en Colab" if EN_COLAB else "Corriendo fuera de Colab: algunas celdas solo muestran un aviso")
```

## 1. Escribir en el disco de la máquina (temporal)

```python
from pathlib import Path

notas = Path("notas.txt")
notas.write_text("Esto vive en la máquina de Colab: se pierde cuando la máquina se borra.\n", encoding="utf-8")
print("Escrito:", notas.resolve())
print(notas.read_text(encoding="utf-8"))
```

## 2. Bajar y subir archivos

`google.colab.files` abre el diálogo del navegador. También puedes usar el panel 📁 **Archivos** de la izquierda.

```python
if EN_COLAB:
    from google.colab import files
    files.download("notas.txt")       # el navegador descarga el archivo
else:
    print("files.download solo existe en Colab")
```

```python
if EN_COLAB:
    from google.colab import files
    subidos = files.upload()          # elige uno o más archivos de tu computador
    for nombre, contenido in subidos.items():
        print(f"Subido {nombre}: {len(contenido)} bytes")
else:
    print("files.upload solo existe en Colab")
```

## 3. Conectar tu Google Drive

`drive.mount` pide **tu permiso** (una ventana de Google) y deja tu Drive como una carpeta más. Lo que guardes en
`/content/drive/MyDrive` queda en **tu** Drive, aunque la máquina se borre.

> Colab pide acceso a **tu** Drive, no al de MiAtlas.cl: el curso nunca ve tus archivos.

```python
if EN_COLAB:
    from google.colab import drive
    drive.mount("/content/drive")
    CARPETA = Path("/content/drive/MyDrive/MiAtlas-Colab")
else:
    CARPETA = Path("MiAtlas-Colab")   # fuera de Colab, una carpeta local
CARPETA.mkdir(parents=True, exist_ok=True)
print("Carpeta de trabajo:", CARPETA)
```

```python
guardado = CARPETA / "notas.txt"
guardado.write_text(notas.read_text(encoding="utf-8").replace("se pierde", "NO se pierde (está en Drive)"), encoding="utf-8")
print("Guardado en:", guardado)
```

## 4. Guardar el cuaderno

  Google** (FAQ).
- Este cuaderno lo abriste desde GitHub: para conservar tus cambios usa **Archivo → Guardar una copia en Drive**.
- **Archivo → Guardar una copia en GitHub** lo guarda en un repositorio tuyo (Colab te pide autorizar GitHub).
- Cualquier cuaderno público de GitHub se abre con la dirección
  `https://colab.research.google.com/github/<usuario>/<repositorio>/blob/<rama>/<ruta>.ipynb`.

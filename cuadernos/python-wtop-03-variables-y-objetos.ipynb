## Las variables de Python son punteros

Asignar variables en Python es tan fácil como poner el nombre de la variable a la izquierda del signo igual (`=`):

```python
# asignar 4 a la variable x
x = 4
```

Parece sencillo, pero si tienes un modelo mental equivocado de lo que hace esta operación, el funcionamiento de
Python te puede parecer confuso. Veámoslo brevemente.

En muchos lenguajes de programación conviene pensar las variables como recipientes o baldes donde pones datos. En C,
por ejemplo, cuando escribes

```C
// código C
int x = 4;
```

estás definiendo un «balde de memoria» llamado `x` y poniendo el valor `4` dentro. En Python, en cambio, conviene
pensar las variables no como recipientes sino como punteros. Así, cuando en Python escribes

```python
x = 4
```

estás definiendo un *puntero* llamado `x` que apunta a otro balde que contiene el valor `4`. Fíjate en una
consecuencia: como las variables de Python solo apuntan a distintos objetos, no hace falta «declarar» la variable, ¡y
ni siquiera que la variable apunte siempre a información del mismo tipo! En este sentido se dice que Python tiene
*tipado dinámico*: los nombres de variables pueden apuntar a objetos de cualquier tipo. En Python puedes hacer cosas
como esta:

````python
x = 1          # x es un entero
x = 'hola'     # ahora x es una cadena
x = [1, 2, 3]  # ahora x es una lista
````

Quienes usan lenguajes con tipado estático pueden echar de menos la seguridad de tipos que dan declaraciones como
las de C,

```C
int x = 4;
```

pero este tipado dinámico es una de las cosas que hacen que Python sea tan rápido de escribir y fácil de leer.

> 📝 **Nota de MiAtlas.cl:** desde Python 3.5 existen las **anotaciones de tipo** (`x: int = 4`, `def f(n: int) -> str: ...`). Python no las
> hace cumplir al ejecutar: sirven como documentación y para que herramientas como `mypy` o tu editor detecten errores
> antes de ejecutar. El tipado sigue siendo dinámico.

Este enfoque de «variable como puntero» tiene una consecuencia que debes conocer. Si dos nombres de variable apuntan
al mismo objeto *mutable*, ¡al cambiar uno también cambia el otro! Por ejemplo, creemos y modifiquemos una lista:

````python
x = [1, 2, 3]
y = x
````

Creamos dos variables, `x` e `y`, que apuntan al mismo objeto. Por eso, si modificamos la lista a través de uno de
sus nombres, veremos que la «otra» lista también cambia:

````python
print(y)
````

````python
x.append(4)  # agregar 4 a la lista a la que apunta x
print(y)     # ¡la lista de y también cambió!
````

Este comportamiento puede parecer confuso si piensas, erróneamente, que las variables son baldes que contienen datos.
Pero si piensas, correctamente, que las variables son punteros a objetos, tiene todo el sentido.

Fíjate también en que si usamos «`=`» para asignar otro valor a `x`, el valor de `y` no se ve afectado: la
asignación solo cambia el objeto al que apunta la variable:

````python
x = 'otra cosa'
print(y)  # y no cambia
````

> 📝 **Nota de MiAtlas.cl:** si necesitas una lista **independiente**, haz una copia: `y = x.copy()`, `y = list(x)` o `y = x[:]`. El operador
> `is` te dice si dos nombres apuntan al **mismo** objeto, y `==` si los valores son **iguales**:

````python
a = [1, 2, 3]
b = a          # mismo objeto
c = a.copy()   # copia independiente

print(b is a, c is a)  # True False
print(c == a)          # True: mismos valores
````

De nuevo, esto tiene todo el sentido si piensas en `x` e `y` como punteros, y en el operador «`=`» como una operación
que cambia aquello a lo que apunta el nombre.

Podrías preguntarte si esta idea de punteros hace difícil seguir las operaciones aritméticas en Python, pero Python
está pensado para que eso no sea un problema. Los números, las cadenas y otros *tipos simples* son inmutables: no
puedes cambiar su valor, solo puedes cambiar a qué valores apuntan las variables. Por eso es totalmente seguro hacer
operaciones como esta:

````python
x = 10
y = x
x += 5  # sumar 5 al valor de x y asignarlo a x
print("x =", x)
print("y =", y)
````

Al ejecutar `x += 5` no modificamos el valor del objeto `10` al que apunta `x`; lo que hacemos es cambiar la variable
`x` para que apunte a un nuevo objeto entero con valor `15`. Por eso el valor de `y` no se ve afectado por la
operación.

## Todo es un objeto

Python es un lenguaje de programación orientado a objetos, y en Python todo es un objeto.

Veamos qué significa esto. Antes vimos que las variables son simplemente punteros, y que los nombres de variables no
llevan información de tipo. Esto lleva a algunos a afirmar, erróneamente, que Python es un lenguaje sin tipos. ¡Pero
no es así! Mira lo siguiente:

````python
x = 4
type(x)
````

````python
x = 'hola'
type(x)
````

````python
x = 3.14159
type(x)
````

Python sí tiene tipos; pero los tipos no están ligados a los nombres de las variables, sino *a los objetos mismos*.

En los lenguajes orientados a objetos como Python, un *objeto* es una entidad que contiene datos junto con metadatos
y funcionalidades asociadas. En Python todo es un objeto, lo que significa que toda entidad tiene metadatos (llamados
*atributos*) y funcionalidades asociadas (llamadas *métodos*). A los atributos y métodos se accede con la sintaxis del
punto.

Por ejemplo, antes vimos que las listas tienen un método `append`, que agrega un elemento a la lista, y al que se
accede con el punto («`.`»):

````python
L = [1, 2, 3]
L.append(100)
print(L)
````

Es esperable que objetos compuestos como las listas tengan atributos y métodos, pero lo que a veces sorprende es que
en Python incluso los tipos simples los tienen. Por ejemplo, los tipos numéricos tienen los atributos `real` e `imag`,
que devuelven la parte real y la imaginaria del valor, visto como número complejo:

````python
x = 4.5
print(x.real, "+", x.imag, 'i')
````

Los métodos son como los atributos, salvo que son funciones que puedes llamar con paréntesis de apertura y cierre.
Por ejemplo, los números de punto flotante tienen un método llamado `is_integer` que revisa si el valor es entero:

````python
x = 4.5
x.is_integer()
````

````python
x = 4.0
x.is_integer()
````

Cuando decimos que en Python todo es un objeto, de verdad queremos decir que *todo* es un objeto: incluso los
atributos y métodos de los objetos son a su vez objetos, con su propia información de `type`:

````python
type(x.is_integer)
````

Veremos que esta decisión de diseño de Python, que todo sea un objeto, permite construcciones muy cómodas en el
lenguaje.

## Pruébalo tú

Tienes la lista `original`. Guarda en `resultado` una **copia independiente** de ella y agrega el número `4` a la
copia, de modo que `original` siga siendo `[1, 2, 3]`. Si escribes `resultado = original`, ¿qué pasa con `original`?

````python
original = [1, 2, 3]

resultado = None  # reemplaza None por una copia independiente de original

if resultado is not None:
    resultado.append(4)
print("original:", original)
````

````python
# Comprobación
if resultado is None:
    print("Aún no lo resuelves.")
elif resultado == [1, 2, 3, 4] and original == [1, 2, 3] and resultado is not original:
    print("✅ ¡Correcto!")
else:
    print("❌ Revisa tu respuesta")
````

**Objetivo:** ejecutar tus primeras celdas y entender cómo Colab recuerda lo que ya ejecutaste.

Según su propia documentación, Colab es un servicio de cuadernos Jupyter alojado por Google: no hay que instalar nada y
es **gratis**, con acceso a recursos de cómputo que incluyen **GPU y TPU** ([FAQ de Colab](https://research.google.com/colaboratory/faq.html)).

Un cuaderno tiene dos tipos de celdas:

- **Texto** (como esta): explica, con formato.
- **Código**: Python que se ejecuta en una máquina de Google, no en tu computador.

## 1. Ejecutar una celda

Pon el cursor en la celda de abajo y presiona **Ctrl + Enter** (en Mac, **⌘ + Enter**), o el botón ▶ de su izquierda.
**Shift + Enter** la ejecuta y pasa a la siguiente.

La primera vez Colab tarda unos segundos: está **conectándose a una máquina** (el «entorno de ejecución»).

```python
print("¡Hola desde Colab!")
2 + 3   # la última expresión de la celda se muestra sola
```

## 2. El cuaderno recuerda lo que ejecutaste

Las variables viven en el entorno de ejecución, **no en el orden en que aparecen las celdas**. Ejecuta estas dos en
orden:

```python
nombre = "Ada"
```

```python
print(f"Hola, {nombre}")
```

El número entre corchetes a la izquierda (`[1]`, `[2]`…) dice **en qué orden** se ejecutó cada celda.

> **Experimenta:** reinicia la sesión (menú **Entorno de ejecución** → **Reiniciar sesión**) y ejecuta **solo** la
> celda del `print`. Aparece `NameError`: la variable se perdió, porque reiniciar borra la memoria del entorno.
> Moraleja: un cuaderno debe funcionar ejecutándolo **de arriba abajo** (**Entorno de ejecución → Ejecutar todas**).

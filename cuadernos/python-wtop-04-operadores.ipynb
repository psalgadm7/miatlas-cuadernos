## Operaciones aritméticas
Python implementa siete operadores aritméticos binarios básicos, dos de los cuales también sirven como operadores
unarios. Se resumen en la siguiente tabla:

| Operador     | Nombre           | Descripción                                             |
|--------------|------------------|---------------------------------------------------------|
| ``a + b``    | Suma             | Suma de ``a`` y ``b``                                   |
| ``a - b``    | Resta            | Diferencia entre ``a`` y ``b``                          |
| ``a * b``    | Multiplicación   | Producto de ``a`` y ``b``                               |
| ``a / b``    | División real    | Cociente de ``a`` y ``b``                               |
| ``a // b``   | División entera  | Cociente de ``a`` y ``b``, sin la parte fraccionaria    |
| ``a % b``    | Módulo           | Resto entero de dividir ``a`` por ``b``                 |
| ``a ** b``   | Potencia         | ``a`` elevado a ``b``                                   |
| ``-a``       | Negación         | El opuesto de ``a``                                     |
| ``+a``       | Más unario       | ``a`` sin cambios (casi no se usa)                      |

Estos operadores se pueden usar y combinar de forma intuitiva, con paréntesis para agrupar las operaciones.
Por ejemplo:

```python
# suma, resta y multiplicación
(4 + 8) * (6.5 - 3)
```

La división entera es la división real sin la parte fraccionaria:

```python
# División real
print(11 / 2)
```

```python
# División entera
print(11 // 2)
```

Por último, hay un octavo operador aritmético, que se agregó en Python 3.5: el operador ``a @ b``, pensado
para indicar el *producto matricial* de ``a`` y ``b`` en los paquetes de álgebra lineal.

> 📝 **Nota de MiAtlas.cl:** el original comentaba aquí cómo dividía Python 2; lo quitamos porque Python 2 ya
> no tiene soporte desde 2020. En Python 3, ``/`` siempre da un número decimal y ``//`` la división entera.
> Los números de Python no implementan ``@``: lo vas a usar con NumPy (por ejemplo, ``A @ B`` multiplica dos
> matrices).

## Operaciones bit a bit
Además de las operaciones numéricas habituales, Python tiene operadores para hacer operaciones lógicas *bit a bit*
sobre enteros. Se usan mucho menos que las operaciones aritméticas, pero conviene saber que existen.
Los seis operadores bit a bit se resumen en esta tabla:

| Operador     | Nombre                  | Descripción                                    |
|--------------|-------------------------|------------------------------------------------|
| ``a & b``    | Y bit a bit (AND)       | Bits presentes en ``a`` y en ``b``             |
| <code>a &#124; b</code>| O bit a bit (OR)        | Bits presentes en ``a``, en ``b`` o en ambos   |
| ``a ^ b``    | O exclusivo (XOR)       | Bits presentes en ``a`` o en ``b``, pero no en ambos |
| ``a << b``   | Desplazamiento a la izquierda | Desplaza los bits de ``a`` ``b`` posiciones a la izquierda |
| ``a >> b``   | Desplazamiento a la derecha   | Desplaza los bits de ``a`` ``b`` posiciones a la derecha   |
| ``~a``       | Negación bit a bit (NOT)| Invierte los bits de ``a``                     |

Estos operadores solo tienen sentido si pensamos en la representación binaria de los números, que puedes ver con la
función integrada ``bin``:

```python
bin(10)
```

El resultado empieza con ``'0b'``, que indica una representación binaria.
El resto de los dígitos dice que el número 10 es la suma $1 \cdot 2^3 + 0 \cdot 2^2 + 1 \cdot 2^1 + 0 \cdot 2^0$.
Del mismo modo, podemos escribir:

```python
bin(4)
```

Ahora, con el O bit a bit, obtenemos el número que combina los bits de 4 y de 10:

```python
4 | 10
```

```python
bin(4 | 10)
```

Los operadores bit a bit no son tan útiles en el día a día como los aritméticos, pero vale la pena verlos al
menos una vez para entender qué tipo de operación hacen.
En particular, quienes vienen de otros lenguajes a veces caen en la tentación de usar XOR (``a ^ b``) cuando en
realidad quieren una potencia (``a ** b``).

## Operaciones de asignación
Ya vimos que las variables se asignan con el operador «``=``» y que el valor queda guardado para usarlo después.
Por ejemplo:

```python
a = 24
print(a)
```

Podemos usar estas variables en expresiones con cualquiera de los operadores anteriores.
Por ejemplo, para sumarle 2 a ``a`` escribimos:

```python
a + 2
```

Quizás queramos actualizar la variable ``a`` con ese nuevo valor; en ese caso podríamos combinar la suma y la
asignación y escribir ``a = a + 2``.
Como esta combinación de operación y asignación es tan común, Python trae operadores de actualización para todas las
operaciones aritméticas:

```python
a += 2  # equivale a a = a + 2
print(a)
```

Hay un operador de asignación aumentada para cada uno de los operadores binarios vistos antes; en resumen:

|||||
|-|-|-|-|
|``a += b``| ``a -= b``|``a *= b``| ``a /= b``|
|``a //= b``| ``a %= b``|``a **= b``|``a &= b``|
|<code>a &#124;= b</code>| ``a ^= b``|``a <<= b``| ``a >>= b``|

Cada uno equivale a la operación correspondiente seguida de una asignación: para cualquier operador «``■``», la
expresión ``a ■= b`` equivale a ``a = a ■ b``, con un pequeño matiz.
Con objetos mutables, como listas, arreglos o *DataFrames*, estas asignaciones aumentadas son sutilmente distintas de
su versión larga: modifican el contenido del objeto original en vez de crear un objeto nuevo para guardar el
resultado.

> 📝 **Nota de MiAtlas.cl:** desde Python 3.8 existe además el **operador morsa** ``:=``, que asigna un valor
> *dentro* de una expresión. Por ejemplo, ``if (n := len(datos)) > 10:`` guarda el largo en ``n`` y lo compara en
> la misma línea. Úsalo con moderación: el ``=`` de siempre suele ser más claro.

## Operaciones de comparación

Otro tipo de operación muy útil es la comparación de valores.
Para esto, Python implementa los operadores de comparación habituales, que devuelven los valores booleanos ``True``
(verdadero) y ``False`` (falso). Las operaciones de comparación son estas:

| Operación     | Descripción                          || Operación     | Descripción                          |
|---------------|--------------------------------------||---------------|--------------------------------------|
| ``a == b``    | ``a`` igual a ``b``                  || ``a != b``    | ``a`` distinto de ``b``              |
| ``a < b``     | ``a`` menor que ``b``                || ``a > b``     | ``a`` mayor que ``b``                |
| ``a <= b``    | ``a`` menor o igual que ``b``        || ``a >= b``    | ``a`` mayor o igual que ``b``        |

Estos operadores de comparación se pueden combinar con los aritméticos y los bit a bit para expresar una variedad
casi ilimitada de pruebas sobre los números.
Por ejemplo, podemos saber si un número es impar comprobando que su módulo 2 da 1:

```python
# 25 es impar
25 % 2 == 1
```

```python
# ¿66 es impar?
66 % 2 == 1
```

Podemos encadenar varias comparaciones para comprobar relaciones más complejas:

```python
# comprueba si a está entre 15 y 30
a = 25
15 < a < 30
```

Y, solo para que te duela un poco la cabeza, mira esta comparación:

```python
-1 == ~0
```

Recuerda que ``~`` es el operador que invierte los bits, y por lo visto, al invertir todos los bits del cero
se obtiene -1.
Si te da curiosidad el porqué, busca el esquema de codificación de enteros llamado *complemento a dos*, que es el que
usa Python para representar los enteros con signo, y piensa qué pasa al invertir todos los bits de un entero
codificado así.

## Operaciones booleanas
Cuando trabajas con valores booleanos, Python ofrece operadores para combinarlos según los conceptos habituales de
«y», «o» y «no».
Como era de esperar, estos operadores se escriben con las palabras en inglés ``and``, ``or`` y ``not``:

```python
x = 4
(x < 6) and (x > 2)
```

```python
(x > 10) or (x % 2 == 0)
```

```python
not (x < 6)
```

Quienes aman el álgebra de Boole notarán que no hay un operador XOR; por supuesto, se puede construir de varias
formas combinando los otros operadores.
Igual, hay un truco ingenioso para hacer el XOR de dos valores booleanos:

```python
# (x > 1) xor (x < 10)
(x > 1) != (x < 10)
```

Este tipo de operaciones booleanas será muy útil cuando veamos las *instrucciones de control de flujo*, como
los condicionales y los bucles.

Algo que a veces confunde del lenguaje es cuándo usar los operadores booleanos (``and``, ``or``, ``not``) y cuándo
los bit a bit (``&``, ``|``, ``~``).
La respuesta está en sus nombres: los operadores booleanos se usan cuando quieres calcular el *valor booleano (verdad
o falsedad) de expresiones completas*.
Los operadores bit a bit se usan cuando quieres *operar sobre los bits o componentes individuales de los objetos*.

## Operadores de identidad y de pertenencia

Al igual que ``and``, ``or`` y ``not``, Python tiene operadores que se leen casi como prosa para comprobar la
identidad y la pertenencia. Son los siguientes:

| Operador       | Descripción                                            |
|----------------|--------------------------------------------------------|
| ``a is b``     | Verdadero si ``a`` y ``b`` son el mismo objeto         |
| ``a is not b`` | Verdadero si ``a`` y ``b`` no son el mismo objeto      |
| ``a in b``     | Verdadero si ``a`` es un elemento de ``b``             |
| ``a not in b`` | Verdadero si ``a`` no es un elemento de ``b``          |

### Operadores de identidad: «``is``» e «``is not``»

Los operadores de identidad, «``is``» e «``is not``», comprueban la *identidad de los objetos*.
La identidad es distinta de la igualdad, como se ve aquí:

```python
a = [1, 2, 3]
b = [1, 2, 3]
```

```python
a == b
```

```python
a is b
```

```python
a is not b
```

¿Cómo se ven dos objetos idénticos? Aquí tienes un ejemplo:

```python
a = [1, 2, 3]
b = a
a is b
```

La diferencia entre los dos casos es que en el primero ``a`` y ``b`` apuntan a *objetos distintos*, mientras
que en el segundo apuntan al *mismo objeto*.
Como vimos en el capítulo anterior, las variables de Python son punteros. El operador «``is``» comprueba si las dos
variables apuntan al mismo contenedor (objeto), no si lo que contienen es igual.
Con esto en mente, casi siempre que alguien que empieza siente la tentación de usar «``is``», lo que en realidad
quiere es ``==``.

> 📝 **Nota de MiAtlas.cl:** el uso típico de ``is`` es comparar con ``None``: escribe ``x is None`` o
> ``x is not None``. Desde Python 3.8, comparar con ``is`` contra un número o una cadena escrita en el código
> (por ejemplo, ``x is 1``) produce un aviso ``SyntaxWarning``, justamente porque casi siempre es un error.

### Operadores de pertenencia
Los operadores de pertenencia comprueban si algo es un elemento de un objeto compuesto.
Por ejemplo, podemos escribir:

```python
1 in [1, 2, 3]
```

```python
2 not in [1, 2, 3]
```

Estas operaciones de pertenencia son un ejemplo de lo que hace a Python tan fácil de usar en comparación con
lenguajes de más bajo nivel, como C.
En C, la pertenencia se determinaría, en general, escribiendo a mano un bucle que recorra la lista y compare cada
valor.
En Python basta con escribir lo que quieres saber, de una forma que recuerda a la prosa en inglés.

## Introducción

Python nació a fines de la década de 1980 como un lenguaje para enseñar y para escribir *scripts*. Desde entonces se
convirtió en una herramienta esencial para muchas personas que programan, para ingenieros, investigadores y
científicos de datos, tanto en la academia como en la industria.

El autor del libro, Jake VanderPlas, es astrónomo y se dedica a crear y difundir herramientas libres para la ciencia
con muchos datos. Para él, Python calza casi a la perfección con los problemas de cada día: sacar sentido de grandes
conjuntos de datos astronómicos, extraer y ordenar datos de la Web o automatizar tareas de investigación.

El atractivo de Python está en su simpleza y su belleza, y también en el gran ecosistema de herramientas
especializadas construidas sobre él. Gran parte del código Python de computación científica y ciencia de datos se
apoya en unos pocos paquetes maduros:

- [NumPy](https://numpy.org): almacena y calcula de forma eficiente con arreglos de datos de varias dimensiones.
- [SciPy](https://scipy.org): reúne herramientas numéricas, como integración e interpolación.
- [Pandas](https://pandas.pydata.org): ofrece el objeto `DataFrame` y métodos para manipular, filtrar, agrupar y
  transformar datos.
- [Matplotlib](https://matplotlib.org): crea gráficos y figuras con calidad de publicación.
- [Scikit-Learn](https://scikit-learn.org): un conjunto uniforme de herramientas para aplicar los algoritmos más
  comunes de aprendizaje automático.
- [IPython/Jupyter](https://jupyter.org): una terminal mejorada y un entorno de cuadernos interactivos, útil para el
  análisis exploratorio y para crear documentos ejecutables. El libro original se escribió entero en cuadernos Jupyter.

Hay muchos paquetes más: si quieres hacer una tarea científica o de análisis de datos, lo más probable es que alguien
ya haya escrito un paquete que la hace.

## Cómo ejecutar código Python

Python es un lenguaje flexible, y hay varias formas de usarlo según la tarea. Algo que lo distingue de otros
lenguajes es que es *interpretado* y no *compilado*: se ejecuta línea por línea, y eso permite programar de forma
interactiva, algo que no es posible de manera directa en lenguajes compilados como Fortran, C o Java.

Veremos cuatro formas principales de ejecutar código Python: el *intérprete de Python*, el *intérprete IPython*,
los *scripts independientes* y el *cuaderno Jupyter*.

### El intérprete de Python

La forma más básica de ejecutar código Python es línea por línea dentro del *intérprete de Python*. Para iniciarlo,
instala Python y escribe `python` en la línea de comandos (la Terminal en macOS y Linux, o el Símbolo del sistema o
PowerShell en Windows):

```
$ python3
Python 3.12.x (...)
Type "help", "copyright", "credits" or "license" for more information.
>>>
```

Con el intérprete abierto puedes escribir y ejecutar fragmentos de código. Aquí lo usamos como una calculadora
sencilla, haciendo cálculos y asignando valores a variables:

```python
>>> 1 + 1
2
>>> x = 5
>>> x * 3
15
```

El intérprete es muy cómodo para probar fragmentos pequeños de código y experimentar con secuencias cortas de
operaciones.

> 📝 **Nota de MiAtlas.cl:** el original mostraba Python 3.5.1 de Anaconda. Hoy, en Linux y macOS el comando suele ser `python3`; en Windows,
> el instalador de python.org trae el lanzador `py` (por ejemplo, `py` o `py -3.12`). Para salir del intérprete,
> escribe `exit()`.

### El intérprete IPython

Si pasas mucho tiempo en el intérprete básico de Python, notarás que le faltan muchas funciones de un entorno de
desarrollo interactivo completo. Una alternativa es *IPython* (por *Interactive Python*), un intérprete con muchas
mejoras. Se inicia escribiendo `ipython` en la línea de comandos:

```
$ ipython
Python 3.12.x (...)
Type 'copyright', 'credits' or 'license' for more information
IPython 8.x -- An enhanced Interactive Python. Type '?' for help.

In [1]:
```

La diferencia visible entre el intérprete de Python y el de IPython está en el indicador: Python usa `>>>`, mientras
que IPython numera las entradas (por ejemplo, `In [1]:`). Igual que antes, podemos ejecutar código línea por línea:

```ipython
In [1]: 1 + 1
Out[1]: 2

In [2]: x = 5

In [3]: x * 3
Out[3]: 15
```

Fíjate en que, así como se numeran las entradas, también se numera la salida de cada comando. IPython ofrece muchas
funciones útiles; el capítulo 16 del libro sugiere dónde aprender más.

> 📝 **Nota de MiAtlas.cl:** el original decía que IPython viene incluido en Anaconda. En tu computador se instala con `pip install ipython`.
> IPython es además el motor que ejecuta el código en Jupyter y en Colab, por eso allí funcionan sus comandos
> especiales, como `%timeit` o `objeto?` para ver la ayuda.

### Scripts de Python independientes

Ejecutar fragmentos línea por línea sirve en algunos casos, pero para programas más complejos conviene guardar el
código en un archivo y ejecutarlo todo de una vez. Por convención, los scripts de Python se guardan en archivos con
extensión *.py*. Por ejemplo, creemos un script llamado *prueba.py* con este contenido:

```python
# archivo: prueba.py
print("Ejecutando prueba.py")
x = 5
print("El resultado es", 3 * x)
```

Para ejecutarlo, nos aseguramos de que esté en la carpeta actual y escribimos `python` *`nombre_del_archivo`* en la
línea de comandos:

```
$ python3 prueba.py
Ejecutando prueba.py
El resultado es 15
```

Para programas más complejos, crear scripts independientes como este es indispensable.

> 📝 **Nota de MiAtlas.cl:** puedes probarlo sin salir del cuaderno. La primera celda crea el archivo *prueba.py* en la carpeta de trabajo del
> cuaderno, y la segunda lo ejecuta con `runpy`, un módulo de la biblioteca estándar que funciona en Colab y en
> JupyterLite. En Colab también puedes escribir `!python prueba.py` en una celda: el signo `!` envía la línea a la
> terminal de la máquina (esto no funciona en JupyterLite).

````python
codigo = """# archivo: prueba.py
print("Ejecutando prueba.py")
x = 5
print("El resultado es", 3 * x)
"""

with open("prueba.py", "w", encoding="utf-8") as archivo:
    archivo.write(codigo)

print("Archivo prueba.py creado")
````

````python
import runpy

_ = runpy.run_path("prueba.py")
````

### El cuaderno Jupyter

Un híbrido útil entre la terminal interactiva y el script independiente es el *cuaderno Jupyter*: un formato de
documento que combina código ejecutable, texto con formato, gráficos e incluso elementos interactivos en un solo
documento. Aunque el cuaderno nació solo para Python, hoy es compatible con muchos lenguajes de programación y es una
parte esencial del [Proyecto Jupyter](https://jupyter.org/). El cuaderno sirve como entorno de desarrollo y también
para compartir trabajo mediante relatos que mezclan código, figuras, datos y texto.

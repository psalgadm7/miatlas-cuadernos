**Objetivo:** cargar una tabla, explorarla y hacer gráficos.

Usamos los **pingüinos de Palmer**: 344 pingüinos de tres especies, medidos en la Antártica entre 2007 y 2009.
Los datos son de dominio público (**CC0**) y los publicó Allison Horst en el paquete `palmerpenguins`.

```python
import pandas as pd
from pathlib import Path

URL = "https://raw.githubusercontent.com/allisonhorst/palmerpenguins/main/inst/extdata/penguins.csv"
local = Path("penguins.csv")            # si el archivo ya está junto al cuaderno, se usa ese
pinguinos = pd.read_csv(local if local.exists() else URL)
print(pinguinos.shape[0], "filas y", pinguinos.shape[1], "columnas")
pinguinos.head()
```

## 1. Explorar

```python
pinguinos.info()
```

```python
pinguinos.describe().round(1)
```

```python
# ¿Cuántos datos faltan en cada columna?
pinguinos.isna().sum()
```

```python
pinguinos["species"].value_counts()
```

## 2. Agrupar

```python
# Promedios por especie (sin contar los datos faltantes)
pinguinos.groupby("species")[["bill_length_mm", "flipper_length_mm", "body_mass_g"]].mean().round(1)
```

## 3. Gráficos con matplotlib

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(7, 4))
for especie, grupo in pinguinos.groupby("species"):
    ax.hist(grupo["flipper_length_mm"].dropna(), bins=20, alpha=0.6, label=especie)
ax.set_xlabel("Largo de la aleta (mm)")
ax.set_ylabel("Pingüinos")
ax.set_title("Largo de la aleta por especie")
ax.legend()
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(6, 5))
for especie, grupo in pinguinos.groupby("species"):
    ax.scatter(grupo["bill_length_mm"], grupo["bill_depth_mm"], s=15, label=especie)
ax.set_xlabel("Largo del pico (mm)")
ax.set_ylabel("Alto del pico (mm)")
ax.set_title("Cada especie forma su propio grupo")
ax.legend()
fig.savefig("pico.png", dpi=120, bbox_inches="tight")   # queda en la máquina: bájalo o guárdalo en Drive (tema 2)
plt.show()
```

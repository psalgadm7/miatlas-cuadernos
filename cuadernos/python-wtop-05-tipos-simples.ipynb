## Enteros
El tipo numérico más básico es el entero.
Cualquier número sin punto decimal es un entero:

```python
x = 1
type(x)
```

Los enteros de Python son bastante más sofisticados que los de lenguajes como ``C``.
Los enteros de C tienen precisión fija y suelen desbordarse al llegar a cierto valor (a menudo cerca de $2^{31}$ o
$2^{63}$, según el sistema).
Los enteros de Python tienen precisión variable, así que puedes hacer cálculos que en otros lenguajes se
desbordarían:

```python
2 ** 200
```

Otra característica práctica de los enteros de Python es que, por defecto, la división convierte el resultado
a punto flotante:

```python
5 / 2
```

Si quieres una división entera, que descarta la parte decimal, usa el operador de división entera:

```python
5 // 2
```

> 📝 **Nota de MiAtlas.cl:** el original comparaba aquí con Python 2 (que truncaba la división de enteros y
> tenía dos tipos, ``int`` y ``long``). Lo quitamos: en Python 3 hay un solo tipo ``int``, de precisión variable.
> Dos detalles actuales:
>
> - Puedes separar los dígitos con guiones bajos para leer mejor los números grandes: ``1_000_000`` es lo mismo
>   que ``1000000``.
> - Desde Python 3.11, convertir a texto un entero de más de 4300 dígitos (por ejemplo, con ``print``) lanza un
>   ``ValueError`` como medida de seguridad; el límite se cambia con ``sys.set_int_max_str_digits()``. Los cálculos
>   no tienen ese límite.

```python
poblacion = 19_000_000
print(poblacion, poblacion == 19000000)
```

## Números de punto flotante
El tipo de punto flotante guarda números con parte fraccionaria.
Se pueden escribir con la notación decimal habitual o con notación exponencial:

```python
x = 0.000005
y = 5e-6
print(x == y)
```

```python
x = 1400000.00
y = 1.4e6
print(x == y)
```

En la notación exponencial, la ``e`` o ``E`` se lee «…por diez elevado a…»,
de modo que ``1.4e6`` se interpreta como $1.4 \times 10^6$.

Un entero se puede convertir explícitamente en número de punto flotante con el constructor ``float``:

```python
float(1)
```

### Un paréntesis: la precisión del punto flotante
Algo que debes tener presente con la aritmética de punto flotante es que su precisión es limitada, lo que puede hacer
que las comparaciones de igualdad no sean fiables. Por ejemplo:

```python
0.1 + 0.2 == 0.3
```

¿Por qué pasa esto? No es algo propio de Python: se debe al formato de precisión fija con que se guardan los
números de punto flotante en binario, que usan la mayoría (si no todas) de las plataformas de cálculo científico.
Todos los lenguajes de programación que usan punto flotante los guardan en una cantidad fija de bits, y eso hace que
algunos números solo se puedan representar de forma aproximada.
Lo podemos ver si imprimimos los tres valores con mucha precisión:

```python
print(f"0.1 = {0.1:.17f}")
print(f"0.2 = {0.2:.17f}")
print(f"0.3 = {0.3:.17f}")
```

> 📝 **Nota de MiAtlas.cl:** el original usaba ``"0.1 = {0:.17f}".format(0.1)``. Hoy la forma recomendada de
> dar formato es con *f-strings* (``f"{valor:.17f}"``), que existen desde Python 3.6; ``.format()`` sigue
> funcionando y da el mismo resultado.

Estamos acostumbrados a pensar los números en notación decimal (base 10), en la que cada fracción se expresa
como una suma de potencias de 10:
$$
1 /8 = 1\cdot 10^{-1} + 2\cdot 10^{-2} + 5\cdot 10^{-3}
$$
En la conocida representación en base 10, lo escribimos con la expresión decimal $0.125$.

Los computadores suelen guardar los valores en notación binaria, en la que cada número se expresa como una suma de
potencias de 2:
$$
1/8 = 0\cdot 2^{-1} + 0\cdot 2^{-2} + 1\cdot 2^{-3}
$$
En base 2, esto se escribe $0.001_2$, donde el subíndice 2 indica notación binaria.
El valor $0.125 = 0.001_2$ resulta ser un número que tanto la notación binaria como la decimal pueden representar con
una cantidad finita de dígitos.

En la representación en base 10 seguramente conoces números que no se pueden expresar con una cantidad finita de
dígitos. Por ejemplo, al dividir $1$ por $3$ se obtiene, en notación decimal:
$$
1 / 3 = 0.333333333\cdots
$$
Los 3 siguen para siempre: para representar ese cociente con exactitud, ¡harían falta infinitos dígitos!

Del mismo modo, hay números cuya representación binaria requiere infinitos dígitos.
Por ejemplo:
$$
1 / 10 = 0.00011001100110011\cdots_2
$$
Así como la notación decimal necesita infinitos dígitos para representar $1/3$ con exactitud, la notación binaria
necesita infinitos dígitos para representar $1/10$.
En la mayoría de los sistemas, Python corta internamente estas representaciones 52 bits después del primer bit
distinto de cero.

Este error de redondeo es un mal necesario al trabajar con números de punto flotante.
La mejor forma de lidiar con él es recordar siempre que la aritmética de punto flotante es aproximada y *nunca*
confiar en comparaciones de igualdad exacta entre valores de punto flotante.

> 📝 **Nota de MiAtlas.cl:** para comparar decimales, usa ``math.isclose``, que comprueba si dos valores son
> iguales dentro de una tolerancia (por defecto, relativa de ``1e-09``):

```python
import math
math.isclose(0.1 + 0.2, 0.3)
```

## Números complejos
Los números complejos tienen una parte real y una imaginaria (ambas de punto flotante).
Ya vimos los enteros y los números reales; con ellos podemos construir un número complejo:

```python
complex(1, 2)
```

También podemos usar el sufijo «``j``» en las expresiones para indicar la parte imaginaria:

```python
1 + 2j
```

Los números complejos tienen varios atributos y métodos interesantes, que mostramos brevemente aquí:

```python
c = 3 + 4j
```

```python
c.real  # parte real
```

```python
c.imag  # parte imaginaria
```

```python
c.conjugate()  # conjugado
```

```python
abs(c)  # módulo, es decir, sqrt(c.real ** 2 + c.imag ** 2)
```

## Cadenas de texto
En Python, las cadenas se crean con comillas simples o dobles:

```python
mensaje = "dime qué te gusta"
respuesta = 'queso'
```

Python tiene muchísimas funciones y métodos de cadenas muy útiles; aquí van algunos:

```python
# largo de la cadena
len(respuesta)
```

```python
# pasar a mayúsculas; ver también str.lower()
respuesta.upper()
```

```python
# primera letra en mayúscula; ver también str.title()
mensaje.capitalize()
```

```python
# concatenación con +
mensaje + respuesta
```

```python
# la multiplicación es una concatenación repetida
5 * respuesta
```

```python
# acceder a caracteres individuales (la indexación parte en cero)
mensaje[0]
```

Para ver más sobre la indexación en Python, revisa la sección «Listas» del capítulo 06 (estructuras de datos).

> 📝 **Nota de MiAtlas.cl:** para armar textos con valores dentro, la forma recomendada son las
> ***f-strings***: una cadena con ``f`` delante, en la que todo lo que va entre llaves se evalúa como expresión.
> Después de ``:`` puedes dar un formato (``.2f`` = dos decimales). El método ``.format()`` es la alternativa
> anterior y sigue funcionando. Desde Python 3.12, dentro de las llaves de una f-string puedes usar incluso el mismo
> tipo de comillas que la rodea.

```python
precio = 1990.5
print(f"La {respuesta} tiene {len(respuesta)} letras y cuesta ${precio:.2f}")
print("La {} tiene {} letras y cuesta ${:.2f}".format(respuesta, len(respuesta), precio))
```

## El tipo ``None``
Python incluye un tipo especial, ``NoneType``, que tiene un solo valor posible: ``None``. Por ejemplo:

```python
type(None)
```

Verás ``None`` en muchos lugares, pero quizás su uso más común es como valor que devuelve una función por
defecto. Por ejemplo, la función ``print()`` no devuelve nada, pero igual podemos capturar su valor:

```python
valor_devuelto = print('abc')
```

```python
print(valor_devuelto)
```

Del mismo modo, cualquier función de Python que no devuelve un valor en realidad devuelve ``None``.

## El tipo booleano
El tipo booleano es un tipo simple con dos valores posibles, ``True`` y ``False``, y es lo que devuelven los
operadores de comparación que vimos antes:

```python
resultado = (4 < 5)
resultado
```

```python
type(resultado)
```

Ten en cuenta que los valores booleanos distinguen mayúsculas de minúsculas: a diferencia de otros lenguajes,
¡``True`` y ``False`` se escriben con mayúscula inicial!

```python
print(True, False)
```

Los booleanos también se pueden construir con el constructor ``bool()``: los valores de cualquier otro tipo se
convierten en booleanos según reglas predecibles.
Por ejemplo, cualquier tipo numérico es ``False`` si es igual a cero, y ``True`` en caso contrario:

```python
bool(2014)
```

```python
bool(0)
```

```python
bool(3.1415)
```

La conversión a booleano de ``None`` siempre es ``False``:

```python
bool(None)
```

Para las cadenas, ``bool(s)`` es ``False`` si la cadena está vacía y ``True`` en cualquier otro caso:

```python
bool("")
```

```python
bool("abc")
```

Para las secuencias, que veremos en el próximo capítulo, la representación booleana es ``False`` si la
secuencia está vacía y ``True`` en cualquier otro caso:

```python
bool([1, 2, 3])
```

```python
bool([])
```

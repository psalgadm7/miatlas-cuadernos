## 1. La máquina que te tocó

```python
import os, shutil

print("Núcleos de CPU:", os.cpu_count())
try:
    with open("/proc/meminfo") as f:
        total_kb = int(f.readline().split()[1])
    print(f"Memoria RAM: {total_kb / 1024**2:.1f} GB")
except OSError:
    print("Memoria RAM: no se pudo leer /proc/meminfo en este sistema")
disco = shutil.disk_usage("/")
print(f"Disco: {disco.free / 1024**3:.0f} GB libres de {disco.total / 1024**3:.0f} GB")
```

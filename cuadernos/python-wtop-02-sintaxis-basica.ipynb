Python se desarrolló en sus inicios como un lenguaje para enseñar, pero su facilidad de uso y su sintaxis limpia
hicieron que lo adoptaran tanto principiantes como expertos. La limpieza de su sintaxis llevó a algunos a llamarlo
«pseudocódigo ejecutable», y la experiencia del autor es que suele ser mucho más fácil leer y entender un script de
Python que uno parecido escrito, por ejemplo, en C. Aquí empezaremos a revisar las principales características de la
sintaxis de Python.

La *sintaxis* es la estructura del lenguaje, es decir, qué forma un programa bien escrito. Por ahora no nos
centraremos en la *semántica* (el significado de las palabras y símbolos dentro de la sintaxis): volveremos a ella
más adelante.

Mira el siguiente ejemplo de código:

````python
# fijar el punto medio
punto_medio = 5

# crear dos listas vacías
menores = []; mayores = []

# separar los números en menores y mayores
for i in range(10):
    if (i < punto_medio):
        menores.append(i)
    else:
        mayores.append(i)

print("menores:", menores)
print("mayores:", mayores)
````

Este script es algo tonto, pero muestra de forma compacta varios aspectos importantes de la sintaxis de Python.
Recorrámoslo y comentemos algunas de sus características sintácticas.

## Los comentarios se marcan con `#`

El script empieza con un comentario:

```python
# fijar el punto medio
```

En Python los comentarios se indican con el signo numeral (`#`), y el intérprete ignora todo lo que viene después de
él en la línea. Así, puedes tener comentarios solos en una línea, como el anterior, y también comentarios al final de
una instrucción. Por ejemplo:

```python
x += 2  # forma abreviada de x = x + 2
```

Python no tiene una sintaxis para comentarios de varias líneas, como `/* ... */` en C y C++, aunque a menudo se usan
cadenas de varias líneas en su reemplazo (más sobre esto en el capítulo 14, «Manipulación de cadenas y expresiones
regulares»).

## El fin de línea termina una instrucción

La siguiente línea del script es

```python
punto_medio = 5
```

Es una asignación: creamos una variable llamada `punto_medio` y le asignamos el valor `5`. Fíjate en que el final de
la instrucción lo marca simplemente el final de la línea. Esto es distinto de lenguajes como C y C++, donde cada
instrucción debe terminar con punto y coma (`;`).

En Python, si quieres que una instrucción siga en la línea siguiente, puedes usar el marcador «`\`»:

````python
x = 1 + 2 + 3 + 4 +\
    5 + 6 + 7 + 8
````

También se puede continuar una expresión en la línea siguiente dentro de paréntesis, sin usar el marcador «`\`»:

````python
x = (1 + 2 + 3 + 4 +
     5 + 6 + 7 + 8)
````

La mayoría de las guías de estilo de Python recomiendan la segunda forma de continuar líneas (dentro de paréntesis)
antes que la primera (con el marcador «`\`»).

## El punto y coma puede terminar una instrucción (opcional)

A veces conviene poner varias instrucciones en una sola línea. La siguiente parte del script es

```python
menores = []; mayores = []
```

Aquí se ve cómo el punto y coma (`;`), conocido en C, puede usarse de forma opcional en Python para poner dos
instrucciones en una sola línea. En la práctica equivale por completo a escribir

```python
menores = []
mayores = []
```

La mayoría de las guías de estilo de Python desaconsejan usar el punto y coma para poner varias instrucciones en una
línea, aunque a veces resulta cómodo.

## Sangría: ¡los espacios en blanco importan!

Ahora llegamos al bloque principal de código:

```python
for i in range(10):
    if i < punto_medio:
        menores.append(i)
    else:
        mayores.append(i)
```

Es una instrucción compuesta de control de flujo, con un bucle y una condición; veremos este tipo de instrucciones
más adelante. Por ahora, fíjate en que muestra la que quizás es la característica más polémica de la sintaxis de
Python: ¡los espacios en blanco tienen significado!

En los lenguajes de programación, un *bloque* de código es un conjunto de instrucciones que se trata como una unidad.
En C, por ejemplo, los bloques se marcan con llaves:

```C
// código C
for(int i=0; i<100; i++)
   {
      // las llaves indican el bloque de código
      total += i;
   }
```

En Python, los bloques de código se marcan con la *sangría*:

```python
for i in range(100):
    # la sangría indica el bloque de código
    total += i
```

En Python, un bloque con sangría siempre va precedido por dos puntos (`:`) al final de la línea anterior.

La sangría ayuda a mantener el estilo uniforme y legible que a muchos les gusta del código Python. Pero puede
confundir a quien recién empieza; por ejemplo, estos dos fragmentos dan resultados distintos:

```python
>>> if x < 4:         >>> if x < 4:
...     y = x * 2     ...     y = x * 2
...     print(x)      ... print(x)
```

En el fragmento de la izquierda, `print(x)` está dentro del bloque con sangría y solo se ejecuta si `x` es menor que
`4`. En el de la derecha, `print(x)` está fuera del bloque, ¡y se ejecuta sea cual sea el valor de `x`!

El uso de espacios con significado suele sorprender a quienes vienen de otros lenguajes, pero en la práctica lleva a
un código mucho más consistente y legible que en lenguajes que no obligan a sangrar los bloques. Si no te gusta, el
autor te anima a darle una oportunidad: como le pasó a él, puede que termines apreciándolo.

Por último, la *cantidad* de espacios que usas para sangrar un bloque la decides tú, siempre que sea la misma en todo
el script. Por convención, la mayoría de las guías de estilo recomiendan sangrar con cuatro espacios, y esa es la
convención que seguimos en estos cuadernos. Muchos editores de texto, como Emacs y Vim, tienen modos para Python que
sangran con cuatro espacios de forma automática.

> 📝 **Nota de MiAtlas.cl:** hoy también lo hacen editores como VS Code. En Colab y en JupyterLite, al escribir `:` y pulsar Enter, la línea
> siguiente queda sangrada sola. Si mezclas tabulaciones y espacios de forma ambigua, Python 3 lanza un `TabError`:
> usa siempre espacios.

## Los espacios *dentro* de las líneas no importan

Aunque el lema de los *espacios con significado* vale para los espacios *al comienzo* de las líneas (que indican un
bloque de código), los espacios *dentro* de una línea de código Python no importan. Por ejemplo, estas tres
expresiones son equivalentes:

````python
x=1+2
x = 1 + 2
x             =        1    +                2
````

Abusar de esta flexibilidad puede dañar la legibilidad del código; de hecho, abusar de los espacios es una de las
formas principales de ofuscar código a propósito (algo que algunos hacen por deporte). Usar bien los espacios hace el
código mucho más legible, sobre todo cuando los operadores van uno tras otro. Compara estas dos expresiones que elevan
a un exponente negativo:

```python
x=10**-2
```

y

```python
x = 10 ** -2
```

La segunda, con espacios, se lee mucho más fácil de un vistazo. La mayoría de las guías de estilo de Python
recomiendan un espacio alrededor de los operadores binarios y ninguno junto a los operadores unarios. Veremos más
sobre los operadores de Python en el capítulo 4, «Semántica básica de Python: operadores».

## Los paréntesis sirven para agrupar o para llamar

En el fragmento anterior vimos dos usos de los paréntesis. Primero, el uso habitual para agrupar instrucciones u
operaciones matemáticas:

````python
2 * (3 + 4)
````

También indican que se está *llamando* a una *función*. En el siguiente fragmento se usa la función `print()` para
mostrar el contenido de una variable (ver el apunte más abajo). La llamada se indica con un par de paréntesis, de
apertura y cierre, con los *argumentos* de la función adentro:

````python
print('primer valor:', 1)
````

````python
print('segundo valor:', 2)
````

Algunas funciones se llaman sin argumentos, y en ese caso igual hay que escribir los paréntesis de apertura y cierre
para indicar que se ejecuta la función. Un ejemplo es el método `sort` de las listas:

````python
L = [4,2,3,1]
L.sort()
print(L)
````

Los «`()`» después de `sort` indican que la función debe ejecutarse, y son obligatorios aunque no haga falta ningún
argumento.

## Apunte: la función `print()`

Más arriba usamos como ejemplo la función `print()`. En Python, `print()` es una **función**: se llama con
paréntesis y recibe como argumentos los valores que quieres mostrar, separados por comas.

```python
>>> print("primer valor:", 1)
primer valor: 1
```

Si ves código con `print` sin paréntesis (`print "primer valor:", 1`), es código antiguo de Python 2, que ya no
funciona en Python 3.

> 📝 **Nota de MiAtlas.cl:** el original dedicaba este apunte a comparar `print` en Python 2 y Python 3; dejamos solo lo vigente. Dos cosas útiles
> de `print()` hoy: los argumentos con nombre `sep` (qué va entre los valores) y `end` (qué va al final), y las
> ***f-strings***, la forma recomendada de armar textos con valores: se antepone una `f` a la cadena y cada valor va
> entre llaves, con un formato opcional después de `:` (por ejemplo, `:.2f` para dos decimales). El método `.format()`
> es una alternativa que verás en código más antiguo.

````python
fruta = "manzana"
cantidad = 3
precio = 1.5

print("a", "b", "c", sep="-", end=".\n")
print(f"{cantidad} unidades de {fruta} cuestan {cantidad * precio:.2f}")
print("{} unidades de {} cuestan {:.2f}".format(cantidad, fruta, cantidad * precio))
````

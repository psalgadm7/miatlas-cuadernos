El *control de flujo* es donde la programación se pone realmente en marcha.
Sin él, un programa es solo una lista de instrucciones que se ejecutan una tras otra.
Con el control de flujo puedes ejecutar ciertos bloques de código de forma condicional o repetida: estas piezas
básicas se combinan para crear programas sorprendentemente sofisticados.

Aquí veremos las *instrucciones condicionales* (`if`, `elif` y `else`) y las *instrucciones de bucle*
(`for` y `while`, junto con `break`, `continue` y `pass`).

## Instrucciones condicionales: `if`-`elif`-`else`

Las instrucciones condicionales, a menudo llamadas instrucciones *si-entonces*, permiten ejecutar ciertas partes del
código según una condición booleana. Un ejemplo básico de condicional en Python es este:

````python
x = -15

if x == 0:
    print(x, "es cero")
elif x > 0:
    print(x, "es positivo")
elif x < 0:
    print(x, "es negativo")
else:
    print(x, "no se parece a nada que haya visto...")
````

Fíjate en especial en el uso de los dos puntos (`:`) y de la sangría para delimitar los bloques de código.

Python adopta el `if` y el `else` de muchos otros lenguajes; su palabra clave más particular es `elif`, una
contracción de «else if». En estas cláusulas condicionales, los bloques `elif` y `else` son opcionales; además,
puedes incluir tantas instrucciones `elif` como quieras, o ninguna.

> 📝 **Nota de MiAtlas.cl:** desde Python 3.10 existe también la instrucción `match`, que compara un valor contra
> varios patrones (parecida al `switch` de otros lenguajes, pero más potente). Para elegir entre unos pocos casos,
> `if`-`elif`-`else` sigue siendo lo más claro. Un ejemplo breve:
>
> ```python
> match comando:
>     case "salir":
>         print("Hasta luego")
>     case "ayuda" | "?":
>         print("Comandos: salir, ayuda")
>     case _:
>         print("No entiendo", comando)
> ```

## Bucles `for`

Los bucles en Python sirven para ejecutar una instrucción una y otra vez.
Por ejemplo, si queremos imprimir cada uno de los elementos de una lista, podemos usar un bucle `for`:

````python
for N in [2, 3, 5, 7]:
    print(N, end=' ')  # imprime todo en la misma línea
````

Fíjate en lo simple que es el bucle `for`: indicamos la variable que queremos usar, la secuencia que queremos
recorrer, y las unimos con el operador `in` de una forma intuitiva y legible.
Más precisamente, el objeto a la derecha del `in` puede ser cualquier *iterable* de Python.
Un iterable se puede pensar como una secuencia generalizada; los veremos en el capítulo 10, *Iteradores*.

Por ejemplo, uno de los iterables más usados en Python es el objeto `range`, que genera una secuencia de números:

````python
for i in range(10):
    print(i, end=' ')
````

Fíjate en que, por omisión, el rango empieza en cero y que, por convención, el extremo superior no se incluye.
Los objetos `range` también admiten valores más elaborados:

````python
# rango de 5 a 10
list(range(5, 10))
````

````python
# rango de 0 a 10 de 2 en 2
list(range(0, 10, 2))
````

Quizás notes que el significado de los argumentos de `range` se parece mucho a la sintaxis de rebanado que vimos con
las listas en el capítulo 6, *Estructuras de datos incorporadas*.

> 📝 **Nota de MiAtlas.cl:** `range()` no crea una lista con todos los números: es un objeto liviano que los entrega
> uno a uno. Por eso, para ver sus valores juntos, aquí lo convertimos con `list(...)`.

## Bucles `while`

El otro tipo de bucle en Python es `while`, que se repite mientras se cumpla una condición:

````python
i = 0
while i < 10:
    print(i, end=' ')
    i += 1
````

La condición del bucle `while` se evalúa como una expresión booleana, y el bucle se ejecuta hasta que esa expresión
resulta `False`.

## `break` y `continue`: ajusta tus bucles

Hay dos instrucciones útiles para ajustar la forma en que se ejecutan los bucles:

- La instrucción `break` sale del bucle por completo.
- La instrucción `continue` se salta el resto de la vuelta actual y pasa a la siguiente.

Ambas sirven tanto en bucles `for` como en bucles `while`.

Este es un ejemplo de `continue` para imprimir los números impares.
En este caso, el resultado se podría lograr igual de bien con un `if`-`else`, pero a veces `continue` es una forma más
cómoda de expresar la idea que tienes en mente:

````python
for n in range(20):
    # si el resto de n / 2 es 0, salta el resto del bucle
    if n % 2 == 0:
        continue
    print(n, end=' ')
````

Este es un ejemplo de `break` en una tarea menos trivial.
El bucle llena una lista con todos los números de Fibonacci hasta cierto valor:

````python
a, b = 0, 1
amax = 100
L = []

while True:
    (a, b) = (b, a + b)
    if a > amax:
        break
    L.append(a)

print(L)
````

Fíjate en que usamos un bucle `while True`, que se repetiría para siempre si no tuviera una instrucción `break`.

> 📝 **Nota de MiAtlas.cl:** la introducción menciona también `pass`, que el original no llega a mostrar.
> `pass` es una instrucción que no hace nada: sirve cuando la sintaxis exige un bloque pero todavía no tienes nada
> que poner en él.
>
> ```python
> for n in range(3):
>     pass  # por ahora, nada
> ```

## Bucles con un bloque `else`

Un patrón poco usado de Python es la instrucción `else` como parte de un bucle `for` o `while`.
Ya vimos el bloque `else`: en un condicional se ejecuta si todas las condiciones `if` y `elif` resultan `False`.
El `else` de los bucles es quizás una de las instrucciones con el nombre más confuso de Python; el autor prefiere
pensarlo como un `nobreak` («sin break»): el bloque `else` se ejecuta solo si el bucle termina de forma natural, sin
encontrar una instrucción `break`.

Como ejemplo de dónde puede ser útil, considera la siguiente implementación (no optimizada) de la *criba de
Eratóstenes*, un algoritmo conocido para encontrar números primos:

````python
L = []
nmax = 30

for n in range(2, nmax):
    for factor in L:
        if n % factor == 0:
            break
    else:  # sin break
        L.append(n)
print(L)
````

La instrucción `else` solo se ejecuta si ninguno de los factores divide al número.
La instrucción `else` funciona de manera similar con el bucle `while`.

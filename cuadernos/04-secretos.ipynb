**Objetivo:** usar una clave (de una API, por ejemplo) **sin escribirla nunca en el cuaderno**.

Un cuaderno se comparte, se sube a GitHub y queda en el historial de Drive. Una clave escrita en una celda la puede
leer cualquiera que vea el cuaderno, y quien la tenga puede gastar a tu nombre.

```python
# ¿Estamos en Colab? Los cuadernos funcionan también fuera de Colab, con avisos en lo que es solo de Colab.
try:
    import google.colab  # noqa: F401
    EN_COLAB = True
except ImportError:
    EN_COLAB = False
print("Corriendo en Colab" if EN_COLAB else "Corriendo fuera de Colab: algunas celdas solo muestran un aviso")
```

## 1. Guardar el secreto en Colab

1. En la barra de la izquierda, abre el panel 🔑 **Secretos**.
2. **Agregar nuevo secreto**: nombre `MI_PRIMER_SECRETO` y un valor de prueba (no uses una clave real).
3. Activa el **acceso del cuaderno** para ese secreto.

El valor queda en tu cuenta de Google, no en el archivo del cuaderno.

## 2. Leerlo desde el código

```python
def leer_secreto_colab(nombre):
    """Devuelve el valor del secreto, o None con un aviso si falta o no tiene acceso."""
    if not EN_COLAB:
        print("Fuera de Colab no hay panel de secretos: usa una variable de entorno (os.environ).")
        return None
    from google.colab import userdata
    try:
        return userdata.get(nombre)
    except userdata.SecretNotFoundError:
        print(f"No existe el secreto {nombre}: créalo en el panel 🔑.")
    except userdata.NotebookAccessError:
        print(f"El secreto {nombre} existe, pero este cuaderno no tiene acceso: actívalo en el panel 🔑.")
    return None


valor = leer_secreto_colab("MI_PRIMER_SECRETO")
if valor is not None:
    print(f"Leído: {len(valor)} caracteres")   # nunca se imprime el valor
```

## 3. Buenas prácticas

- **Nunca** imprimas el valor ni lo escribas en un archivo: muestra, como mucho, su largo.
- Si una clave llegó a un cuaderno compartido, **revócala** en el servicio que la emitió y crea otra.
- Las funciones de IA de Colab (Gemini) **no tienen acceso** a tus secretos por defecto (FAQ).

He intentado cubrir las piezas y los patrones del lenguaje que más le sirven a quien hace ciencia de datos con Python, pero esto no es, ni de lejos, una introducción completa.
Si quieres profundizar en el lenguaje mismo y en cómo usarlo bien, estos son algunos recursos que recomiendo:

- [*Fluent Python*](https://www.oreilly.com/library/view/fluent-python-2nd/9781492056348/), de Luciano Ramalho. Un excelente libro de O'Reilly que explora las buenas prácticas y los modismos de Python, incluido cómo sacarle el máximo provecho a la biblioteca estándar.
- [*Dive Into Python 3*](https://diveintopython3.net/), de Mark Pilgrim. Un libro gratuito en línea que presenta el lenguaje Python desde cero.
- [*Learn Python the Hard Way*](https://learnpythonthehardway.org/), de Zed Shaw. Este libro sigue el enfoque de «aprender probando» y pone énfasis, a propósito, en la que quizás sea la habilidad más útil que puede aprender quien programa: buscar en internet lo que no entiende.
- *Python Essential Reference*, de David Beazley. Este monstruo de 700 páginas está muy bien escrito y cubre prácticamente todo lo que hay que saber sobre el lenguaje Python y sus bibliotecas incorporadas. Para un recorrido más orientado a aplicaciones, revisa el [*Python Cookbook*](https://www.oreilly.com/library/view/python-cookbook-3rd/9781449357337/) de Beazley.

> 📝 **Nota de MiAtlas.cl:** algunos de estos libros cambiaron desde 2016:
>
> - *Fluent Python* tiene una 2.ª edición (2022), que es la del enlace.
> - *Dive Into Python 3* es la versión para Python 3 del libro que citaba el autor; el original era para Python 2, que ya no se usa.
> - El sucesor de *Python Essential Reference* es [*Python Distilled*](https://www.oreilly.com/library/view/python-distilled/9780134173399/) (2021), del mismo autor; su sitio es https://www.dabeaz.com/.
> - Todos estos libros están en inglés.

Para profundizar en las herramientas de Python para ciencia de datos y computación científica, recomiendo estos libros:

- [*Python Data Science Handbook*](https://jakevdp.github.io/PythonDataScienceHandbook/), de este servidor. Este libro empieza justo donde termina este informe y es una guía completa de las herramientas esenciales del ecosistema de ciencia de datos de Python, desde la limpieza y manipulación de datos hasta el aprendizaje automático.
- [*Effective Computation in Physics*](https://www.oreilly.com/library/view/effective-computation-in/9781491901564/), de Katie Huff y Anthony Scopatz, sirve a muchas personas más allá de la investigación en física. Es una introducción paso a paso, desde cero, a la computación científica, que incluye una excelente presentación de muchas de las herramientas mencionadas en este informe.
- [*Python for Data Analysis*](https://wesmckinney.com/book/), de Wes McKinney, creador del paquete Pandas. Este libro trata en detalle la biblioteca Pandas y entrega información útil sobre algunas de las otras herramientas que la hacen posible.

> 📝 **Nota de MiAtlas.cl:** el *Python Data Science Handbook* y *Python for Data Analysis* (3.ª edición) se pueden leer gratis en línea, en los enlaces de arriba.

Por último, para una mirada aún más amplia de lo que hay, recomiendo lo siguiente:

- **Los recursos de Python de O'Reilly:** O'Reilly publica varios libros excelentes sobre Python y sobre temas especializados del mundo Python.
- **PyCon, SciPy y PyData.** Las conferencias [PyCon](https://pycon.org/), [SciPy](https://conference.scipy.org/) y [PyData](https://pydata.org/) reúnen a miles de asistentes cada año, y publican la mayor parte de sus programas como videos gratuitos en línea. > 📝 **Nota de MiAtlas.cl:** el archivo de videos de estas conferencias está reunido en [PyVideo](https://pyvideo.org/). PyCon se organiza en muchos países; en [pycon.org](https://pycon.org/) está la lista de eventos, y en Chile la comunidad [Python Chile](https://pythonchile.cl/).

## Novedades de cada versión

> 📝 **Nota de MiAtlas.cl:** esta sección tampoco está en el original. El libro es de 2016, cuando la versión vigente era Python 3.5. Python publica una versión nueva cada año, y cada una trae su página de novedades («What's New»). Estos cuadernos están actualizados a **Python 3.12**. Este es un resumen de lo más útil para quien aprende, con enlaces a las notas oficiales:

| Versión | Lo más destacado |
|---|---|
| [3.6](https://docs.python.org/es/3/whatsnew/3.6.html) | *f-strings* (PEP 498), separadores ``_`` en números (``1_000_000``) |
| [3.7](https://docs.python.org/es/3/whatsnew/3.7.html) | los diccionarios conservan el orden de inserción (garantizado por el lenguaje), ``dataclasses`` |
| [3.8](https://docs.python.org/es/3/whatsnew/3.8.html) | operador morsa ``:=``, ``f"{x=}"`` para depurar |
| [3.9](https://docs.python.org/es/3/whatsnew/3.9.html) | ``\|`` para unir diccionarios, ``str.removeprefix()`` y ``removesuffix()``, módulo ``zoneinfo`` |
| [3.10](https://docs.python.org/es/3/whatsnew/3.10.html) | ``match``/``case`` (coincidencia de patrones), mensajes de error más claros, ``X \| Y`` en anotaciones de tipo |
| [3.11](https://docs.python.org/es/3/whatsnew/3.11.html) | bastante más rápido, grupos de excepciones (``except*``), módulo ``tomllib`` |
| [3.12](https://docs.python.org/es/3/whatsnew/3.12.html) | *f-strings* más flexibles (PEP 701), nueva sintaxis para parámetros de tipo (PEP 695), mejores mensajes de error |
| [3.13](https://docs.python.org/es/3/whatsnew/3.13.html) | nuevo intérprete interactivo, modo sin GIL (experimental), compilador JIT (experimental) |
| [3.14](https://docs.python.org/es/3/whatsnew/3.14.html) | plantillas de cadena o *t-strings* (PEP 750), evaluación diferida de anotaciones (PEP 649), soporte de Zstandard (PEP 784) |
| [3.15](https://docs.python.org/es/3/whatsnew/3.15.html) | importaciones perezosas explícitas (PEP 810), tipo ``frozendict`` (PEP 814), UTF-8 como codificación por defecto (PEP 686) |

El índice de todas las versiones está en [Novedades de Python](https://docs.python.org/es/3/whatsnew/index.html), y el calendario de soporte de cada versión, en la [guía de desarrollo de Python](https://devguide.python.org/versions/) (en inglés).

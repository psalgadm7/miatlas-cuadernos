## Cargar módulos: la instrucción ``import``

Para cargar módulos incorporados y de terceros, Python ofrece la instrucción ``import``.
Hay varias formas de usarla; las mencionamos brevemente aquí, de la más recomendada a la menos recomendada.

### Importación explícita del módulo

La importación explícita de un módulo conserva su contenido dentro de un espacio de nombres.
Ese espacio de nombres se usa después para referirse a su contenido, con un «``.``» entre ambos.
Por ejemplo, aquí importamos el módulo incorporado ``math`` y calculamos el coseno de pi:

````python
import math
math.cos(math.pi)
````

### Importación explícita del módulo con un alias

Con nombres de módulo largos, no es cómodo escribir el nombre completo cada vez que accedes a algún elemento.
Por eso es habitual usar el patrón «``import ... as ...``» para crear un alias más corto del espacio de nombres.
Por ejemplo, el paquete NumPy (*Numerical Python*), un paquete de terceros muy usado en ciencia de datos, se importa por convención con el alias ``np``:

````python
import numpy as np
np.cos(np.pi)
````

> 📝 **Nota de MiAtlas.cl:** NumPy ya viene instalado en Colab y en JupyterLite. Desde NumPy 2.0 el resultado se muestra como ``np.float64(-1.0)`` y no como ``-1.0``: es el mismo número, solo que ahora la representación dice también su tipo.

### Importación explícita del contenido de un módulo

A veces, en lugar de importar el espacio de nombres del módulo, quieres importar solo algunos elementos concretos.
Eso se hace con el patrón «``from ... import ...``».
Por ejemplo, podemos importar solo la función ``cos`` y la constante ``pi`` del módulo ``math``:

````python
from math import cos, pi
cos(pi)
````

### Importación implícita del contenido de un módulo

Por último, a veces es útil importar todo el contenido del módulo al espacio de nombres local.
Eso se hace con el patrón «``from ... import *``»:

````python
from math import *
sin(pi) ** 2 + cos(pi) ** 2
````

Este patrón debe usarse con moderación, si es que se usa.
El problema es que este tipo de importación puede sobrescribir nombres de funciones que no querías sobrescribir, y como la instrucción es implícita, cuesta saber qué cambió.

Por ejemplo, Python tiene una función incorporada ``sum`` que sirve para varias operaciones:

````python
help(sum)
````

Podemos usarla para calcular la suma de una secuencia partiendo de cierto valor (aquí partiremos de ``-1``):

````python
sum(range(5), -1)
````

Ahora observa qué pasa si hacemos *exactamente la misma llamada* después de importar ``*`` desde ``numpy``:

````python
from numpy import *
````

````python
sum(range(5), -1)
````

¡El resultado no coincide!
La razón es que la instrucción ``import *`` *reemplaza* la función incorporada ``sum`` por la función ``numpy.sum``, que se llama de otra manera: en la primera, sumamos ``range(5)`` partiendo de ``-1``; en la segunda, sumamos ``range(5)`` a lo largo del último eje (indicado por ``-1``).
Este tipo de situación puede aparecer si no tienes cuidado al usar «``import *``»; por eso es mejor evitarlo, a menos que sepas exactamente lo que haces.

> 📝 **Nota de MiAtlas.cl:** para recuperar la ``sum`` original basta con borrar el nombre que la tapaba (``del sum``); Python vuelve a encontrar la función incorporada. También puedes llamarla siempre como ``builtins.sum``. Ojo: ``from numpy import *`` tapó muchos otros nombres (``pi``, ``cos``, ``max``…); si te enredas, reinicia el entorno de ejecución.

````python
del sum
sum(range(5), -1)
````

## Importar desde la biblioteca estándar de Python

La biblioteca estándar de Python contiene muchos módulos incorporados útiles, que puedes conocer a fondo en la [documentación de Python](https://docs.python.org/es/3/library/index.html).
Cualquiera de ellos se puede importar con la instrucción ``import`` y después explorar con la función ``help`` que vimos en la sección anterior.
Esta es una lista muy incompleta de algunos módulos que quizás quieras explorar y aprender:

- ``os`` y ``sys``: herramientas para comunicarse con el sistema operativo, como recorrer la estructura de directorios y ejecutar comandos de la terminal.
- ``math`` y ``cmath``: funciones y operaciones matemáticas con números reales y complejos.
- ``itertools``: herramientas para construir iteradores y generadores, y trabajar con ellos.
- ``functools``: herramientas que ayudan a programar en estilo funcional.
- ``random``: herramientas para generar números pseudoaleatorios.
- ``pickle``: herramientas de persistencia de objetos: guardarlos en disco y cargarlos desde él.
- ``json`` y ``csv``: herramientas para leer archivos en formato JSON y CSV.
- ``urllib``: herramientas para hacer peticiones HTTP y otras peticiones web.

Encontrarás información sobre estos módulos, y muchos más, en la documentación de la biblioteca estándar de Python: https://docs.python.org/es/3/library/index.html.

> 📝 **Nota de MiAtlas.cl:** desde 2016 la biblioteca estándar ganó módulos que hoy se usan a diario: ``pathlib`` (rutas de archivos como objetos), ``statistics`` (media, mediana, desviación…), ``dataclasses`` (clases de datos, desde 3.7), ``zoneinfo`` (zonas horarias, desde 3.9) y ``tomllib`` (leer archivos TOML, desde 3.11). Y un consejo de seguridad: nunca cargues con ``pickle`` un archivo de origen desconocido, porque puede ejecutar código.

````python
import statistics
from pathlib import Path

print(statistics.mean([2, 4, 9]))
print(Path("datos") / "ventas.csv")
````

### Crear tu propio módulo

> 📝 **Nota de MiAtlas.cl:** un módulo no es más que un archivo ``.py``. Si lo guardas en la carpeta donde trabajas, puedes importarlo por su nombre. Aquí lo creamos desde Python para que funcione igual en Colab, en JupyterLite y en tu computador. Una carpeta con varios módulos (y, por convención, un archivo ``__init__.py``) es un **paquete**.

````python
import importlib
from pathlib import Path

Path("saludos.py").write_text(
    '"""Un módulo de ejemplo."""\n'
    "\n"
    "def saludar(nombre):\n"
    "    return f'Hola, {nombre}'\n",
    encoding="utf-8",
)
importlib.invalidate_caches()  # avisa a Python de que hay un archivo nuevo

import saludos
saludos.saludar("mundo")
````

## Importar módulos de terceros

Una de las cosas que hacen útil a Python, sobre todo en el mundo de la ciencia de datos, es su ecosistema de módulos de terceros.
Se importan igual que los módulos incorporados, pero antes hay que instalarlos en tu sistema.
El registro estándar de estos módulos es el *Python Package Index* (*PyPI*), en la web en https://pypi.org/.
Para mayor comodidad, Python incluye un programa llamado ``pip`` (un acrónimo recursivo: «*pip installs packages*», «pip instala paquetes»), que descarga automáticamente los paquetes publicados en PyPI.
Por ejemplo, si quieres instalar el paquete ``supersmoother``, que escribí yo, basta con escribir lo siguiente en la línea de comandos:
```
$ pip install supersmoother
```
El código fuente del paquete se descarga automáticamente del repositorio PyPI, y el paquete se instala en la ruta estándar de Python (siempre que tengas permiso para hacerlo en el computador que usas).

Para más información sobre PyPI y el instalador ``pip``, consulta la documentación en https://pypi.org/ y https://pip.pypa.io/.

> 📝 **Nota de MiAtlas.cl:** así se instala hoy según dónde trabajes.
>
> - **Google Colab:** escribe ``%pip install supersmoother`` en una celda de código. La instrucción mágica ``%pip`` instala en el mismo entorno que usa el cuaderno (es preferible a ``!pip``). Colab ya trae NumPy, pandas, Matplotlib, SciPy y scikit-learn, y lo que instales se pierde al cerrar la sesión: deja la celda de ``%pip`` al comienzo del cuaderno.
> - **JupyterLite:** también acepta ``%pip install``, pero solo para paquetes escritos enteramente en Python o ya preparados para correr en el navegador.
> - **Tu computador:** instala Python desde https://www.python.org/downloads/ y crea un **entorno virtual** por proyecto con ``venv``, para no mezclar paquetes entre proyectos:
>
> ```
> $ python3 -m venv .venv
> $ source .venv/bin/activate        # en Windows: .venv\Scripts\activate
> $ python -m pip install supersmoother
> ```
>
> Más detalles en el tutorial oficial de [entornos virtuales y paquetes](https://docs.python.org/es/3/tutorial/venv.html).
